#### Cooldown

Die Liste a zeigt die Wertentwicklung einer Aktie an aufeinanderfolgenden Tagen. Wir können an einem Tag entweder kaufen, verkaufen oder nichts machen. Wenn wir verkauft haben, können wir nicht am nächsten Tag sofort wieder kaufen (cooldown).
Welcher maximale Gewinn ist möglich?

```python
'''
0 - keine Aktie vorhanden, nicht in cooldown
1 - keine Aktie n, in cooldown
2 - Aktie vorhanden
'''

def dp(i, k):
    if i == n-1:
        if k == 0 or k == 1:
            return 0
        else: return a[i]
    if (i,k) in memo: return memo[(i,k)]
    if k == 0:
        res = max(dp(i+1,0),dp(i+1,2) - a[i])
    elif k == 1:
        res = dp(i+1,0)
    elif k == 2:
        res = max(dp(i+1,2),dp(i+1,1) + a[i])
    memo[(i,k)] = res
    return res


memo = dict()
a = [1,2,3,0,2]
n = len(a)
dp(0,0)
```

Wenn wir die genauen Aktionen wissen wollen, nutzen wir *parent-pointer* im dict *prev*.

```python
'''
0 - keine Aktie vorhanden, nicht in cooldown
1 - keine Aktie n, in cooldown
2 - Aktie vorhanden
'''

def dp(i, k):
    if i == n-1:
        if k == 0 or k == 1:
            prev[(i,k)] = None
            return 0
        else:
            prev[(i,k)] = None
            return a[i]
    if (i,k) in memo: return memo[(i,k)]
    if k == 0:
        res = dp(i+1,0)
        prev[(i,k)] = (i+1,0)
        if dp(i+1,2) - a[i] > res:
            res = dp(i+1,2) - a[i]
            prev[(i,k)] = (i+1,2)
    elif k == 1:
        res = dp(i+1,0)
        prev[(i,k)] = (i+1,0)
    elif k == 2:
        res = dp(i+1,2)
        prev[(i,k)] = (i+1,2)
        if dp(i+1,1) + a[i] > res:
            res =  dp(i+1,1) + a[i] 
            prev[(i,k)] = (i+1,1)
            
       
    memo[(i,k)] = res
    return res

prev = dict()
memo = dict()
a = [48, 12, 60, 93, 97, 42, 25, 64, 17, 56, 85, 93, 9, 48, 52, 42, 58, 85, 81, 84, 69, 36, 1, 54, 23, 15, 72, 15, 11, 94]
n = len(a)
dp(0,0)
```

```python
path = []
state = (0,0)
while state is not None:
    path.append(state)
    state = prev[state]
print(path)
```

```python
actions = []
for i in range(len(path)-1):
    k1 = path[i][1]
    k2 = path[i+1][1]
    if (k1,k2) == (0,2): actions.append('K')
    elif (k1,k2) == (2,1): actions.append('V')
    elif (k1,k2) == (1,0): actions.append('C')
    else: actions.append('x')
if path[-1][1] == 2:
    actions.append('V')
else:
    actions.append('x')
print(actions)
    
```

```python
for x, y in zip(a,actions):
    print(x,y)
```

Die nicht-rekursive Variante:

```python
a = [48, 12, 60, 93, 97, 42, 25, 64, 17, 56, 85, 93, 9, 48, 52, 42, 58, 85, 81, 84, 69, 36, 1, 54, 23, 15, 72, 15, 11, 94]
n = len(a)

dp = [[0]*3 for _ in range(len(a))]
dp[n-1][0] = 0
dp[n-1][1] = 0
dp[n-1][2] = a[n-1]

for i in range(n-2,-1,-1):
    for k in range(3):
        if k == 0:
            res = max(dp[i+1][0],dp[i+1][2] - a[i])
        elif k == 1:
            res = dp[i+1][0]
        elif k == 2:
            res = max(dp[i+1][2],dp[i+1][1] + a[i])

        dp[i][k] = res

dp[0][0]
```
