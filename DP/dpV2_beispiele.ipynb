#### Edit Distance

Gegeben sind zwei Strings Worte w1 und w2 und wir suchen die kürzeste Art, w1 nach w2 zu verwandeln. Jede der folgenden Operationen zählt als ein Schritt:
insert, delete, replace ein Zeichen in x.

<img src='editdistance.png' width='450'>

Wenn unser Problem zwei Listen oder Strings beinhaltet, probieren wir es mit zwei Suffixen.

```python
def dp(i, j):
    '''
    returns: Edit Distance von String w1[i:] zum String w2[j:],  i = 0..len(w1), j = 0..len(w2)
    dp_guess: Was machen wir an der Stelle w1[i]?  Deleten, Inserten, Replacen?
    '''
    if i == n1: return n2-j
    if j == n2: return n1-i

    if (i,j) in memo:
        return memo[(i,j)]

    if w1[i] == w2[j]:         # die zeichen sind gleich
        res = dp(i+1,j+1)     
    else:
        res = 1 + min(dp(i,j+1), dp(i+1,j), dp(i+1,j+1))   # insert
    memo[(i,j)] = res
    return res
```

```python
w1 = 'horse'
w2 = 'ros'

n1 = len(w1)
n2 = len(w2)
memo = dict()

print(dp(0,0))
```

#### Longest Increasing Subsequence

Gegeben eine Liste mit Zahlen. Finde die Länge der längsten Teilfolge, bei der jedes Element größer als sein Vorgänger ist.

<img src='LIS.png' width='600'>

    Versuch:
    dp(i) = Länge der längsten Teilfolge in a[i:]
    dp_guess: Nehmen wir a[i] oder in LIS auf oder nicht?
    

Wenn wir das Element an der Stelle i für LIS nehmen, dann können wir einen der folgenden größeren Elemente als nächstes nehmen (wir kennen also unsere Möglichkeiten), wenn wir aber das Element an Stelle i nicht nehmen, dann kennen wir unsere Möglichkeiten nicht, weil wir das Element, das vorher genommen wurde, nicht kennen. Deswegen formulieren wir unser Problem um:

    dp(i) = Länge der längsten Teilfolge in a[i:], die mit a[i] beginnt.
    dp_guess: Welches Element nehmen wir als nächstes in die LIS auf?

Unser Ausgangsproblem ist dann das Maximum aller dp(i).

```python
def dp(i):
    '''
    returns: Länge der längesten Teilfolge in a[i:], die mit a[i] beginnt,  i=0..n
    '''
    if i in memo: return memo[i]
        
    if i == n: return 0
       
    res = 1 +  max([dp(j) for j in range(i+1,n) if a[i] < a[j]],default = 0)
       
    memo[i] = res
    return res

a = [10,9,2,5,3,7,101,18]
n = len(a)
memo = dict()
max(dp(i) for i in range(n))
```

Laufzeit $O(n^2)$: n subprobleme und pro subproblem O(n) nicht rekursive Arbeit.

#### Longest Common Subsequence

Gegeben zwei Worte. Gesucht ist die längste gemeinsame Subsequence.

    dp(i,j) - Länge der längesten Subsequence der Worte w1[i:] und w2[j:], i=0..len(w1), j=0..len(w2)
    dp_guess:
    wenn w1[i] != w2[j]: 
        Einer von beiden Buchstaben ist definitiv nicht in der LCS. Wir wissen nicht welcher, also probieren wir beide:
            dp(i,j) = max(dp(i,j+1),dp(i+1,j))
    wenn w1[i] == w2[j]: dann nehme ich beide in LCS auf
            dp(i,j) = 1 + dp(i+1,j+1)



```python
def dp(i, j):
    '''
    returns: Länge der längesten Subsequence der Worte w1[i:] und w2[j:], i=0..len(w1), j=0..len(w2)
    '''
    if (i,j) in memo: return memo[(i,j)]
        
    if i == n1 or j == n2: return 0
        
    if w1[i] == w2[j]:
        res = 1 + dp(i+1,j+1)
    else:
        res = max(dp(i,j+1),dp(i+1,j))
        
    memo[(i,j)] = res
    return res
    

w1 = 'HIEROGLYPHOLOGY'
w2 = 'MICHELANGELO'
n1 = len(w1)
n2 = len(w2)
dp(0,0)
```

#### Rod cutting

    
    n = 7
    a = [1,10,13,18,20,31,32] = Preise für Stücke der Länge 1..7

    dp(i) = Maximum-value, wenn die Länge i ist i = 0...n

```python
def dp(i):
    '''
    returns:  Maximum-value, wenn die Länge i ist i = 0...n
    '''
    if i in memo: return memo[i]
        
    if i == 0: return 0
       
    res = max([a[k-1] + dp(i-k) for k in range(1,n+1) if i-k>=0])
       
    memo[i] = res
    return res

a = [1,10,13,18,20,31,32]
memo = dict()
dp(len(a))
```

#### Subset Sum
    

    a = [2,5,7,8,1], T = 21  - 
    n = len(a)

    dp(i,t) = gibt es ein subset in a[i:] mit summe t, i=0..n, t=0..T

    dp_guess: nehme ich a[i] mit in das set oder nicht?

    top-order: abnehmendes i



```python
def dp(i, t):
    '''
    returns: True, falls es ein subset in a[i:] mit summe t gibt, i=0..n, t=0..T 
    '''
    if (i,t) in memo: return memo[(i,t)]
        
    if i == n: 
        if t == 0: return True
        return False

    res = dp(i+1,t) 
    if t-a[i] >= 0:
          res = res or dp(i+1,t-a[i])
        
    memo[(i,t)] = res
    return res

memo = dict()
a = [2,5,7,8,9]
n = len(a) 
T = 21
dp(0,T)
```

Laufzeit: O(n*T) das ist nicht polynomial in input-size.
Input-size ist n+1 (n+1 Worte: die Liste und das T)
Aber wie groß ist T? Wir wissen nicht, wie n und T sich zueinander verhalten. Vielleicht ist es 2^n?
Dann haben wir exponentielle Laufzeit.
Aber solange T nicht wirklich sehr groß ist, ist das ein guter Algorithmus: pseudopolynomial = polynomial in input-size und input-integers.
ist polynomial wenn T polynomial in input_size ist

    dp(i) = Anzahl Möglichkeiten die Zahlen von 1...i in zwei gleiche Summen zu teilen

    dp(i, k) = Anzahl Möglichkeiten mit den Zahlen von 1..i die Summe k zu erreichen

    dp(i, k) = dp(i-1, k-i) + dp(i-1,k)

```python
def dp(i,k):
    '''
    returns: Anzahl Möglichkeiten mit den Zahlen von 1..i die Summe k, k >= 0
    '''
    if (i,k) in memo: return memo[(i,k)]
        
    if i == 1:
        if k <= 1: 
            return 1
        return 0

    res = dp(i-1,k)
    if k-i >= 1:
        res += dp(i-1, k-i)
        
    memo[(i,k)] = res % (10**9
    return res

memo = dict()
n = 7
S = n*(n+1)/2
if S % 2 != 0:
    print(0)
else:
    print(dp(n,S/2))
```
