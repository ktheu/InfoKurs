### Frog 2

(https://atcoder.jp/contests/dp/tasks/dp_b)

Der Frosch sitzt auf der ersten Zahl und will zur letzten hüpfen. Er kann immer nur 1 bis k Zahlen weiter hüpfen. Die Kosten für jeden Sprung ist der Unterschied der Zahlen zwischen denen er springt. Ziel ist es, die Kosten zu minimieren.

<img src='frog2.png' width='600'>

```python
def dp(i):
    '''
    returns: die minimalen Kosten, wenn der Frosch von a[i] bis a[n-1] hüpfen muss
    dp-guess: welches ist der erste Schritt
    '''
    if i in memo: return memo[i]
    if i == n-1: return 0
        
    res = abs(a[i]-a[i+1]) + dp(i+1)              # Sprung 1
    for j in range(2,k+1):                        # Sprung 2 .. k
        if i+j < n and abs(a[i]-a[i+j]) + dp(i+j) < res:
            res = abs(a[i]-a[i+j]) + dp(i+j)
        
    memo[i] = res
    return res

k = 4
a = [40, 10, 20, 70, 80, 10, 20, 70, 80, 60]
n = len(a)
memo = {}
print(dp(0))
```

```python
dp = [0]*len(a)
dp[n-1] = 0
for i in range(n-2,-1,-1):
    res = abs(a[i]-a[i+1]) + dp[i+1]              # Sprung 1
    for j in range(2,k+1):                        # Sprung 2 .. k
        if i+j < n and abs(a[i]-a[i+j]) + dp[i+j] < res:
            res = abs(a[i]-a[i+j]) + dp[i+j]
    dp[i] = res
dp[0]
```

### House Robber

https://leetcode.com/problems/house-robber/description/

Der Räuber will in alle Häuser einer Straße einbrechen. Die zu raubenden Werte sind in einer Liste gegeben. Er darf keine benachbarten Häuser ausrauben. Welchen maximalen Wert kann er erreichen?

<img src='robber1.png' width='500'>


```python
def dp(i):
    if i == n-1:
        return a[i]
    if i == n-2:
        return max(a[-1],a[-2])
    if i in memo: return memo[i]
    res = dp(i+1)
    if a[i] + dp(i+2) > res:
        res = a[i] + dp(i+2)

    memo[i] = res
    return res

a = [2, 7, 9, 6, 1]
n = len(a)
memo = {}
print(dp(0))
```

```python
def dp(i):
    if i >= n:
        return 0
    if i in memo: return memo[i]
    res = dp(i+1)
    if a[i] + dp(i+2) > res:
        res = a[i] + dp(i+2)

    memo[i] = res
    return res

a = [2, 7, 9, 6, 1]
n = len(a)
memo = {}
print(dp(0))
```

#### Minimizing Coins

https://www.cses.fi/problemset/task/1634/

Gegeben sind verschiedene Münzwerte und eine Zahl k. Welches ist die kleinste Anzahl von Münzen, bei der die Summe der Werte genau k ist.
Gib -1 zurück, wenn es nicht geht.

<img src='coins.png' width='401'>

Kleinere Zahlen als das gegebene k sind die kleineren Problemgrößen

<img src='coins2.png' width='601'>




```python
def dp(k):
    '''
    returns: die minimale Anzahl von Münzen, um die Summe k zu erreichen
    dp-guess: welches ist die erste Münze der optimalen Lösung?
    '''
    if k in memo: return memo[k]
    if k == 0: return 0                           # Für die Summe 0 gibt es eine Lösung: keine Münzen
        
    res = float('inf')                            # noch keine Lösung gefunden
    for c in coins:
        if k-c >= 0 and 1 + dp(k-c) < res:
            res = 1 + dp(k-c)
    memo[k] = res
    return res


# coins = [2, 3, 4, 8, 9, 11, 14, 16, 17, 19]
# x = 200
coins = [1, 5, 7]
k = 3
memo = {}
ans = dp(k)
if ans == float('inf'):
    print(-1)
else:
    print(ans)
```
