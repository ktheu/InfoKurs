### House Robber

https://leetcode.com/problems/house-robber/description/

Der Räuber will in alle Häuser einer Straße einbrechen. Die zu raubenden Werte sind in einer Liste gegeben. Er darf keine benachbarten Häuser ausrauben. Welchen maximalen Wert kann er erreichen?

<img src='robber1.png' width='501'>


```python
def dp(i):
    '''
    returns: maximaler Wert für a[i:], i=0..n
    dp_guess: wird a[i] ausgeraubt oder nicht?
    '''
    if i in memo: return memo[i]
        
    if i == n: return 0
    if i == n-1: return a[n-1]
   
    res = max(dp(i+1), a[i]+dp(i+2))

    memo[i] = res
    return res

a = [2, 7, 9, 6, 1]
n = len(a)
memo = {}
print(dp(0))
```

### Climbing Stairs

https://leetcode.com/problems/climbing-stairs/description/

Eine Treppe hat n Stufen. Wir stehen vor der ersten Stufe. Wir können jeweils 1 oder 2 Stufen hoch steigen. Wieviele verschiedene Möglichkeiten gibt es, bis nach ganz oben zu steigen?

<img src='stairs2.png' width='402'>





```python
def dp(i):
    '''
    returns: die Anzahl Möglichkeiten nach oben zu kommen, wenn ich vor Stufe i stehe
    dp_guess: ein Schritt oder zwei Schritte, wenn ich vor Stufe i stehe?
    '''
    if i in memo: return memo[i]
        
    if i == n-1: return 1
    if i == n-2: return 2

    res = dp(i+1) + dp(i+2)
    
    memo[i] = res
    return res

n = 3
memo = dict()
dp(0)
```

#### Minimizing Coins

https://www.cses.fi/problemset/task/1634/

Gegeben sind verschiedene Münzwerte und eine Zahl k. Welches ist die kleinste Anzahl von Münzen, bei der die Summe der Werte genau k ist.
Gib -1 zurück, wenn es nicht geht.

<img src='coins.png' width='401'>

Kleinere Zahlen als das gegebene k sind die kleineren Problemgrößen





```python
def dp(i):
    '''
    returns: die minimale Anzahl von Münzen, um die Summe i zu erreichen, i=0..k
    dp-guess: welches ist die erste Münze einer optimalen Lösung?
    '''
    if i in memo: return memo[i]
        
    if i == 0: return 0                           # Für die Summe 0 gibt es eine Lösung: 0 Münzen 
        
    res = float('inf')                            # noch keine Lösung gefunden
    for c in coins:
        if i-c >= 0 and 1 + dp(i-c) < res:
            res = 1 + dp(i-c)
            
    memo[i] = res
    return res


coins = [1, 5, 7]
k = 3
memo = {}
ans = dp(k)
if ans == float('inf'):
    print(-1)
else:
    print(ans)
```

#### Edit Distance

https://leetcode.com/problems/edit-distance/

Gegeben sind zwei Strings w1 und w2 und wir suchen die kürzeste Art, w1 nach w2 zu verwandeln. Jede der folgenden Operationen zählt als ein Schritt: insert, delete, replace ein Zeichen in w1.

<img src='editdistance.png' width='450'>

```python
def dp(i, j):
    '''
    returns: minimale Anzahl Operationen, um Wort w1[i:] in w2[j:] zu überführen
    dp_guess: was mache ich an der Stelle i in w1? 
    '''

    if (i,j) in memo: return memo[(i,j)]
        
    if i == n1: return n2-j
    if j == n2: return n1-i

    if w1[i] == w2[j]:
        res = dp(i+1,j+1) 
    else:
        res = 1 + min(dp(i,j+1), dp(i+1,j), dp(i+1,j+1))
        
    memo[(i,j)] = res
    return res

w1 = 'horse'
w2 = 'ros'
n1 = len(w1)
n2 = len(w2)
memo = dict()
dp(0,0)
```

#### Longest Increasing Subsequence (LIS)

Gegeben eine Liste mit Zahlen. Finde die Länge der längsten Teilfolge, bei der jedes Element größer als sein Vorgänger ist.

<img src='LIS.png' width='600'>

    Versuch:
    dp(i) = Länge der längsten Teilfolge in a[i:]
    dp_guess: Nehmen wir a[i] oder in LIS auf oder nicht?
    

Wenn wir das Element an der Stelle i für LIS nehmen, dann können wir einen der folgenden größeren Elemente als nächstes nehmen (wir kennen also unsere Möglichkeiten), wenn wir aber das Element an Stelle i nicht nehmen, dann kennen wir unsere Möglichkeiten nicht, weil wir das Element, das vorher genommen wurde, nicht kennen. Deswegen formulieren wir unser Problem um:

    dp(i) = Länge der längsten Teilfolge in a[i:], die mit a[i] beginnt.
    dp_guess: Welches Element nehmen wir als nächstes in die LIS auf?

Unser Ausgangsproblem ist dann das Maximum aller dp(i).

```python
def dp(i):
    '''
    returns: Länge der längesten Teilfolge in a[i:], die mit a[i] beginnt,  i=0..n
    '''
    if i in memo: return memo[i]
        
    if i == n: return 0
       
    res = 1 +  max([dp(j) for j in range(i+1,n) if a[i] < a[j]],default = 0)
       
    memo[i] = res
    return res

a = [10,9,2,5,3,7,101,18]
n = len(a)
memo = dict()
max(dp(i) for i in range(n))
```

Laufzeit $O(n^2)$: n subprobleme und pro subproblem O(n) nicht rekursive Arbeit.

#### Longest Common Subsequence (LCS)

Gegeben zwei Worte. Gesucht ist die Länge der längsten gemeinsamen Subsequence.

<img src='LCS.png' width='600'>

```python
def dp(i, j):
    '''
    returns: Länge der längesten Subsequence der Worte w1[i:] und w2[j:], i=0..n1, j=0..n2
    dp_guess: welcher der Buchstaben w1[i] oder w2[j] wird in die LCS aufgenommen (oder beide oder garkeiner)?
    '''
    if (i,j) in memo: return memo[(i,j)]
        
    if i == n1 or j == n2: return 0
        
    if w1[i] == w2[j]:
        res = 1 + dp(i+1,j+1)
    else:
        res = max(dp(i,j+1),dp(i+1,j))
        
    memo[(i,j)] = res
    return res
    

w1 = 'HIEROGLYPHOLOGY'
w2 = 'MICHELANGELO'
n1 = len(w1)
n2 = len(w2)
dp(0,0)
```

#### Predict the winner

https://leetcode.com/problems/predict-the-winner/description/

2 Spieler nehmen abwechseln von einer Liste vorne oder hinten eine Zahl weg. Gewonnen hat, wer zum Schluss die größte Summe hat. Gib True aus, wenn der Spieler, der beginnt, sicher gewinnen kann. Wenn beide die gleiche Summe haben, hat trotzdem Spieler 1 gewonnen.

<img src='predict_winner.png' width='500'>


    subproblem: Typ substring, d.h. Teilliste von index i bis index j:  a[i:j+1] 

    Bei einem 2player-Spiel empfiehlt es sich, einen weiteren Parameter für den Player einzufügen

    


```python
def dp(i, j, p):
    '''
    returns: maximale Summe für Spieler 1 bei a[i:j+1] wenn Spieler p als erstes dran ist.  i=0..n-1, j=n..i
    p=0: Spieler 1, p=1: Spieler 2
    dp_guess: welche zahl nimmt Spieler p?
    '''
 
    if (i,j,p) in memo: return memo[(i,j,p)]
        
    if i == j: 
        if p == 0: return a[i]
        return 0
        
    if p == 0:
        res = max(a[i]+dp(i+1,j,1), a[j]+dp(i,j-1,1))
    if p == 1:
        res  = min(dp(i+1,j,0), dp(i,j-1,0))

    memo[(i,j,p)] = res
    return res

a = [1,5,233,7]
memo = dict()
me = dp(0,len(a)-1,0)
print(me)
print(me >= sum(a) -me)    
```
