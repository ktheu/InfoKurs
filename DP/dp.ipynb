#### Frog 1

https://atcoder.jp/contests/dp/tasks/dp_a

Gegeben eine Liste a:

    a = [30, 10, 60, 10, 60, 50]

Der Frosch sitzt auf der ersten Zahl und will zur letzten hüpfen. Er kann immer nur 1 oder 2 Zahlen weiter hüpfen. Die Kosten für jeden Sprung ist der Unterschied der Zahlen zwischen denen er springt. Ziel ist es, die Kosten zu minimieren.

<img src='frog1.png' width='500'>

Bei der Rekursion dürfen wir für die Lösung des Problems die Antworten für kleinere Problemgrößen benutzen. 
Weil wir nicht wissen, wie groß der erste Sprung für eine optimale Lösung ist, probieren wir alle möglichen Sprünge aus.

```python
def dp(i):
    '''
    returns: die minimalen Kosten für a[i:]. i=0..n-1
    dp-guess: wie groß war der erste Sprung?
    '''
    if i in memo: return memo[i]
        
    if i == n-1: return 0
    if i == n-2: return abs(a[n-1]-a[n-2])
 
    res = min(abs(a[i]-a[i+1]) + dp(i+1), abs(a[i]-a[i+2]) + dp(i+2))
       
    memo[i] = res
    return res

a = [30, 10, 60, 10, 60, 50]
n = len(a)
memo = {}
print(dp(0))
```

Wir bilden Subproblems. Im ersten Ansatz ist das immer das Originalproblem nur für kleinere Problemgrößen. Bezieht sich das Problem auf eine Liste a, dann versuchen wir es zunächst mit Suffixen, also Teillisten a[i:]. Wir versuchen mit Rekursion alle Subproblems zu lösen.

    n = len(a)
    dp(i) = Lösung des Problems für a[i:],  i=0..n-1 oder i=0..n
    dp(0) = Lösung des Originalproblems

Für die Rekursionsgleichung müssen wir dp(i) ausdrücken mit dp(k)-Werten wobei k größer als i ist: das entspricht kleineren Problemgrößen. 

    dp(i) = .... dp(i+1) ... dp(i+2)       

In vielen Fällen können wir eine Rekursionsgleichung nicht ohne weiteres aufstellen. Dann fragen wir uns: was würden wir von der Lösung dp(i) gerne wissen (dp-guess)? Bei Suffixen bezieht sich die Frage meist auf das, was zu Beginn gemacht wird. Die Frage sollte so gestellt sein, dass wir mit einer Antwort rekursiv weitermachen können. Dann machen wir einen brute force über alle möglichen Antworten und suchen uns das Beste raus.

```python
def dp(i):
    '''
    returns: die minimalen Kosten für a[i:]. 
    dp-guess: wie groß war der erste Sprung?
    '''
    if i in memo: return memo[i]
        
    if i == n-1: return 0
    if i == n-2: return abs(a[n-1]-a[n-2])
 
    res = min(abs(a[i]-a[i+1]) + dp(i+1), abs(a[i]-a[i+2]) + dp(i+2))
       
    memo[i] = res
    return res

a = [30, 10, 60, 10, 60, 50]
n = len(a)
memo = {}
print(dp(0))
```

    Laufzeit = Anzahl Subproblems * nicht-rekursive Arbeit pro Subproblem  
    Die rekursiven Aufrufe dürfen wir wegen der Memoization mit O(1) ansetzen.  
Laufzeit = $O(n)$


#### Frog 2

(https://atcoder.jp/contests/dp/tasks/dp_b)

Der Frosch sitzt auf der ersten Zahl und will zur letzten hüpfen. Er kann immer nur 1 bis k Zahlen weiter hüpfen. Die Kosten für jeden Sprung ist der Unterschied der Zahlen zwischen denen er springt. Ziel ist es, die Kosten zu minimieren.

<img src='frog2.png' width='600'>

```python
def dp(i):
    '''
    returns: die minimalen Kosten für a[i:]. 
    dp_guess: wie groß war der erste Sprung?
    '''
    if i in memo: return memo[i]
        
    if i == n-1: return 0

    res = abs(a[i]-a[i+1]) + dp(i+1)        # erster Sprung 1
    for j in range(2,k+1):                  # gehts mit ersten Sprung 2..k besser?     
        if i+j < n and abs(a[i]-a[i+j]) + dp(i+j) < res:
            res = abs(a[i]-a[i+j]) + dp(i+j)
        
    memo[i] = res
    return res

k = 4
a = [40, 10, 20, 70, 80, 10, 20, 70, 80, 60]
n = len(a)
memo = {}
print(dp(0))
```

Laufzeit = $O(k*n)$

#### Bowling

Gegeben n Bowling-Kegel, wir können bei einem Wurf nur 0,1 oder 2 Kegel treffen. Kegel i hat den Wert a[i]. Wenn man 1 Kegel trifft, dann bekommt man a[i] Punkte, wenn man 2 Kegel trifft, bekommt man das Produkt der Punkte. Man muss nicht alle treffen. Ziel ist es, die Punkte zu maximieren.

<img src='bowling.png' width='601'>

```python
def dp(i):
    '''
    returns: maximale Punktzahl für die Liste a[i:],   i = 0...n
    dp-guess: was wird mit dem ersten Kegel gemacht: stehen gelassen, einzeln oder im Paar getroffen?
    '''
    if i in memo: return memo[i]
        
    if i == n: return 0
    if i == n-1: return max(a[i],0)

    res = max(dp(i+1), a[i]+dp(i+1), a[i]*a[i+1]+ dp(i+2))
    
    memo[i] = res
    return res

a = [1, 1, 9, 9, 2, -5, -5] 
n = len(a)
memo = dict()

print(dp(0))
```

Laufzeit = $O(n)$

#### Cooldown

Die Liste a zeigt die Wertentwicklung einer Aktie an aufeinanderfolgenden Tagen. Wir können an einem Tag entweder kaufen, verkaufen oder nichts machen. Wenn wir verkauft haben, können wir am nächsten Tag nicht sofort wieder kaufen (cooldown).
Welcher maximale Gewinn ist möglich?

<img src='cooldown.png' width='500'>

    dp(i) =  maximaler Gewinn für a[i:]
    
dp-guess: was machen wir am Tag i? Kaufen, verkaufen oder nichts tun? Um zu entscheiden welche Möglichkeiten wir haben, führen wir den zusätzlichen Parameter k ein, der angibt, in welchem Zustand wir uns befinden.



```python
def dp(i, k):
    '''
    returns: maximaler Gewinn für a[i:], wenn zu Beginn im Zustand k, i = 0...n
        k = 0: keine Aktie vorhanden, nicht in cooldown
        k = 1: keine Aktie vorhanden, im cooldown
        k = 2: Aktie vorhanden

    dp-guess: Was machen wir am Tag i, wenn wir uns im Zustand k befinden?
    '''
    if (i,k) in memo: return memo[(i,k)]
        
    if i == n: return 0

    if k == 0:
        res = max(dp(i+1,0),dp(i+1,2) - a[i])
    elif k == 1:
        res = dp(i+1,0)
    elif k == 2:
        res = max(dp(i+1,2),dp(i+1,1) + a[i])
        
    memo[(i,k)] = res
    return res


memo = dict()
a = [1,2,3,0,2]
n = len(a)
dp(0,0)
```

Laufzeit = $O(n)$

#### Vacation

(https://atcoder.jp/contests/dp/tasks/dp_c)

Gegeben sind drei Listen a, b und c der Länge n. Für jeden Index der Liste darf eine Zahl aus einer Liste mit diesem Index gewählt werden. Es dürfen nicht zweimal nacheinander Zahlen aus derselben Liste gewählt werden. Was ist die maximale Summe der gewählten Zahlen? 

<img src='vacation.png' width='400'>

    dp(i) = die maximale Summe für a[i:] 

Wenn wir an Stelle i beginnen, können wir jede der drei Listen wählen, aber wir müssen dafür sorgen, dass die rekursiven Aufrufe nicht dieselbe Liste wieder wählen. Deswegen geben wir als Parameter mit, welche Liste gerade gewählt wurde.

    dp(i, k) = die maximale Summe, wenn wir vom Index i beginnen und dort aus Liste k wählen.
               (k = 0: Liste a, k = 1: Liste b, k = 2, Liste c)

Unser Ausgangsproblem ist dann das Maximum von dp(0,0), dp(0,1), dp(0,2). 

```python
def dp(i,k):
    '''
    returns: maximale Summe, wenn wir vom Index i beginnen und dort aus Liste k wählen. i=1..n
    a: k = 0, b: k = 1, c: k = 2
    dp_guess: welche Liste wird als nächstes gewählt?
    '''
    if (i,k) in memo:  return memo[(i,k)]
        
    if i == n: return 0

    if k == 0:
        res = a[i] + max(dp(i+1,1), dp(i+1,2))
    if k == 1:
        res = b[i] + max(dp(i+1,0), dp(i+1,2))
    if k == 2:
        res = c[i] + max(dp(i+1,0), dp(i+1,1))
        
    memo[(i,k)] = res
    return res
```

```python
a = [10, 20, 30]
b = [40, 50, 60]
c = [70, 80, 90]

n = len(a)
memo = {}
ans = max(dp(0, 0), dp(0, 1), dp(0, 2))
print(ans)
```

Laufzeit = $O(n)$
