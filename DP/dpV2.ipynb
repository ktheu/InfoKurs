Dynamisches Programmieren können wir häufig beschreiben als eine Verbindung von Rekursion, Memoization und ein bisschen brute force.

Nehmen wir an, wir haben eine Problemstellung, die sich auf eine Liste von Daten bezieht. Dann sollten wir uns fragen, ob es uns helfen würde, wenn wir (via Rekursion) die Antworten für kleinere Problemgrößen wüßten. Wenn wir zu dem Schluß kommen, das würde uns schon helfen, aber ganz könnten wir damit die Sache nicht lösen, weil uns ein Detail fehlt, dann sind wir auf der richtigen Spur. Dann machen wir nämlich über das Detail, was wir nicht wissen, ein brute force und suchen uns die beste Lösung raus.

Suffixe für Listen:


    dp(i) = Lösung des Problems für Liste a[i:]  
    dp(0) = Lösung des Originalproblems

    n = len(a)
    Falls das Problem für die leere Liste eine sinnvolle Lösung hat:
        Rekursionsbremse = dp(n)  
    Sonst:
        Rekursionsbremse = dp(n-1)   

Je nachdem, welche Fälle wir in der Rekursionsbremse abfangen, können wir in der Rekursion Indizes ohne weitere Einschränkungen verwenden.

    i == n:    höchster Index für a in der Rekursion: i
    i == n-1:  höchster Index für a in der Rekursion: i+1
    i == n-2:  höchster Index für a in der Rekursion: i+2

Wenn wir beispielsweise in der Rekursionsbremse den Fall n-1 abfangen und wir wollen in der Rekursions einen Ausdruck a[i+2] verwenden, dann müssen wir explizit sicherstellen dass i+2 ein gültiger Index ist, also i+2 < n.

#### Frog 1

https://atcoder.jp/contests/dp/tasks/dp_a

Gegeben eine Liste a:

    a = [30, 10, 60, 10, 60, 50]

Der Frosch sitzt auf der ersten Zahl und will zur letzten hüpfen. Er kann immer nur 1 oder 2 Zahlen weiter hüpfen. Die Kosten für jeden Sprung ist der Unterschied der Zahlen zwischen denen er springt. Ziel ist es, die Kosten zu minimieren.

<img src='frog1.png' width='500'>

```python
def dp(i):
    '''
    returns: die minimalen Kosten, wenn der Frosch von a[i] bis a[n-1] hüpfen muss
    dp-guess: welches ist der erste Schritt
    '''
    if i == n-1: return 0
    if i in memo: return memo[i]

    res = abs(a[i]-a[i+1]) + dp(i+1)  
    if i+2 < n and abs(a[i]-a[i+2]) + dp(i+2) < res:
        res = abs(a[i]-a[i+2]) + dp(i+2)
        
    memo[i] = res
    return res

a = [30, 10, 60, 10, 60, 50]
n = len(a)
memo = {}
print(dp(0))
```

Die Hüpfpositionen für den optimalen Weg merken wir uns mit *parent-pointern* im dictionary *prev*.
Der parent-pointer für Index i ist der Index des rekursiven Aufrufs, der benutzt wurde, um dp(i) zu berechnen.

```python
def dp(i):
    '''
    returns: die minimalen Kosten, wenn der Frosch von a[i] bis a[n-1] hüpfen muss
    dp-guess: welches ist der erste Schritt
    '''
    if i in memo: return memo[i]
    if i == n-1: 
        prev[i] = None
        return 0
        
    res = abs(a[i]-a[i+1]) + dp(i+1)  
    prev[i] = i+1                           # dp(i+1) wurde benutzt, um dp(i) zu berechnen
    if i+2<n and abs(a[i]-a[i+2]) + dp(i+2) < res:
        res = abs(a[i]-a[i+2]) + dp(i+2)
        prev[i] = i+2                       # dp(i+2) wurde benutzt, um dp(i) zu berechnen
    memo[i] = res
    return res

a = [30, 10, 60, 10, 60, 50]
n = len(a)
memo = {}
prev = {}
print(dp(0))
```

#### Laufzeit:

Allgemein gilt für die DP-Programme: Laufzeit = (Anzahl *subproblems*) mal (nicht-rekursive Arbeit pro subproblem).
Die rekursiven Aufrufe können wir dank memoization mit $O(1)$ ansetzen.

Für Frog1 bedeutet das: $O(n)$, statt $O(2^n)$ bei Ausprobieren des Hüpfens an jeder Stelle.
