#### Frog 2 

(https://atcoder.jp/contests/dp/tasks/dp_b)

Wie Frog 1 nur wird jetzt noch eine maximale Sprungweite vorgegeben

```python
10 4
40 10 20 70 80 10 20 70 80 60
```

```python
def dp(i):
    '''
    returns: die minimalen Kosten, wenn der Frosch von a[i] bis a[n-1] hüpfen muss und
    Hüpflängen 1..k erlaubt sind.
    dp-guess: welches ist der erste Schritt
    '''
    if i in memo: return memo[i]
    if i == n-1: return 0    
    res = min([dp(i+r) + abs(a[i] - a[i+r]) if i+r < n else float("inf") for r in range(1, k+1)]) 
    memo[i] = res
    return res

a = [40, 10, 20, 70, 80, 10, 20, 70, 80, 60]
n = len(a)
k = 4
memo = {}
print(dp(0))
```

#### Dice Combinations

https://www.cses.fi/problemset/task/1633/

Gegeben ist eine Zahl n. Wieviele verschiedene Möglichkeiten gibt es, mit einem Würfel mehrmals zu würfeln und in der Summe n zu erhalten.

```python
def dp(i):
    if i == 0: return 1
    if i in memo: return memo[i]
    res = dp(i-1)
    for j in range(2,7):
        if i - j >= 0 and dp(j) < res:
            res = dp(j)
    memo[i] = res
    return res
        
```

#### Minimizing Coins

https://www.cses.fi/problemset/task/1634/


#include <algorithm>
#include <iostream>
#include <vector>
using namespace std;
 
int main() {
    int n, x;
    cin >> n >> x;
    vector<int> coins(n);
    for (int i = 0; i < n; ++i) {
        cin >> coins[i];
    }
 
    vector<int> dp(x + 1, 10000000);
    dp[0] = 0;
    for (int i = 1; i <= x; ++i) {
        int res = 10000000;
        for (int c : coins) {
            if (i - c >= 0) {
                res = min(res, dp[i - c] + 1);
            }
        }
        dp[i] = res;
    }
 
    if (dp[x] >= 10000000) {
        cout << -1 << endl;
    } else {
        cout << dp[x] << endl;
    }
 
    return 0;
}
