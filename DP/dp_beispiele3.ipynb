#### Vacation

(https://atcoder.jp/contests/dp/tasks/dp_c)

Gegeben sind drei Listen a, b und c der Länge N. Für jeden Index der Liste darf eine Zahl aus einer Liste mit diesem Index gewählt werden. Es dürfen nicht zweimal nacheinander Zahlen aus derselben Liste gewählt werden. Was ist die maximale Summe der gewählten Zahlen? 

<img src='vacation.png' width='400'>

Der übliche Ansatz mit Suffixen:

    dp(i) = die maximale Summe, wenn wir vom Index i beginnen.

Wenn wir an Stelle i sind, wissen wir nicht, was wir als nächstes wählen dürfen. Also führen wir einen weiteren Parameter ein.

    dp(i, k) = die maximale Summe, wenn wir vom Index i beginnen und dort aus Liste k wählen.
               (k = 0: Liste a, k = 1: Liste b, k = 2, Liste c)

Unser Ausgangsproblem ist dann das Maximum von dp(0,0), dp(0,1), dp(0,2). 

```python
def dp(i,k):
    '''
    returns: maximaler Gewinn, wenn ich bei i beginne und dort aus Liste k wähle.
    a: k = 0, b: k = 1, c: k = 2
    '''
    if i == n-1:
        return max(a[i],b[i],c[i])
    if (i,k) in memo:
        return memo[(i,k)]
    if k == 0:
        res = a[i] + max(dp(i+1,1), dp(i+1,2))
    if k == 1:
        res = b[i] + max(dp(i+1,0), dp(i+1,2))
    if k == 2:
        res = c[i] + max(dp(i+1,0), dp(i+1,1))
    memo[(i,k)] = res
    return res

# a = [10, 20, 30]
# b = [40, 50, 60]
# c = [70, 80, 90]

a = [6, 8, 2, 7, 4, 2, 7]
b = [7, 8, 5, 8, 6, 3, 5]
c = [8, 3, 2, 6, 8 ,4, 1]

n = len(a)
memo = {}
ans = max(dp(0, 0), dp(0, 1), dp(0, 2))
print(ans)
```

#### Bowling

Gegeben n Bowling-Kegel, wir können bei einem Wurf nur 0,1 oder 2 Kegel treffen. Kegel i hat den Wert a[i]. Wenn man 1 Kegel trifft, dann bekommt man a[i] Punkte, wenn man 2 Kegel trifft, bekommt man das Produkt der Punkte. Man muss nicht alle treffen. Ziel ist es, die Punkte zu maximieren.

<img src='bowling.png' width='601'>

Der übliche Ansatz mit Suffixen:

    dp(i) die maximale Punktzahl, die ich bekommen kann, wenn ich bei Index i beginne.

Wir gehen alle Möglichkeiten durch, wie wir mit dem ersten Kegel bei Index i verfahren können: nicht treffen, einzeln treffen und im Paar treffen. 
Wir wählen die beste Möglichkeit.

```python
def dp(i):
    '''
    returns: maximale Punktzahl, die ich bekommen kann, wenn ich bei Index i beginne.
    '''
    if i == n-1:
        return max(0,a[i])
    if i == n-2:
        return max(0,a[i],a[i+1],a[i]*a[i+1])
    if i in memo:
        return memo[i]
    res = max(dp(i+1), a[i]+dp(i+1), a[i]*a[i+1]+dp(i+2))
    memo[i] = res
    return res

a = [1, 1, 9, 9, 2, -5, -5]
n = len(a)
memo = dict()

print(dp(0))
```

#### Edit Distance

Gegeben sind zwei Strings Worte w1 und w2 und wir suchen die kürzeste Art, w1 nach sy zu verwandeln. Jede der folgenden Operationen zählt als ein Schritt:
insert, delete, replace ein Zeichen in x.

<img src='editdistance.png' width='450'>

Strings können wir behandeln wie Listen von Zeichen. Wir beginnen auch hier mit Suffixen. Da wir allerdings zwei Strings haben, haben wir es mit den Kombinationen von zwei Suffixen zu tun.

    dp(i, j) = Edit Distance von String w1[i:] zum String w2[j:]

Wir müssen jetzt eine Rekursionsbeziehung finden, bei der in den rekursiven Aufrufen auf der rechten Seite mindestens ein Index größer als i bzw. j ist. Wie immer fragen wir uns, was können wir mit dem ersten Zeichen machen, also mit dem Zeichen von w1 an der Stelle i. Wir können es replacen, deleten oder einen insert durchführen. Wir suchen uns aus den Möglichkeiten die beste aus. 

```python
def dp(i, j):
    '''
    returns: minimale Anzahl Operationen, um Wort w1[i:] in w2[j:] zu überführen
    '''
    if len(w1) == 0: return len(w2)
    if len(w2) == 0: return len(w1)
    if i == n1-1:
        if w1[i] in w2[j:]:
            memo[(i,j)] = len(w2)-j-1
            return len(w2)-j-1
        else:
            memo[(i,j)] = len(w2)-j
            return len(w2)-j
    if j == n2-1:
        if w2[j] in w1[i:]:
            memo[(i,j)] = len(w1)-i-1
            return len(w1)-i-1
        else:
            memo[(i,j)] = len(w1)-i
            return len(w1)-i

    if (i,j) in memo:
        return memo[(i,j)]
    if w1[i] == w2[j]:
        res = dp(i+1,j+1) 
    else:
        res = 1 + min(dp(i,j+1), dp(i+1,j+1))
    memo[(i,j)] = res
    return res

w1 = 'horse'
w2 = 'ros'

w1 = 'intention'
w2 = 'execution'

w1 = ''
w2 = ''

w1 = 'park'
w2 = 'spake'
 
n1 = len(w1)
n2 = len(w2)
n = len(a)
memo = dict()

print(dp(0,0))
    
```
