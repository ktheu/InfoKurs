### Git Commands

```python
# Wenn du einen neuen Branch pushst, der auf Remote noch nicht existiert, wird automatisch -u gesetzt (Upstream eingerichtet).
git config --global push.autoSetupRemote simple    

# Remote-Branch wird erstellt und lokaler Branch feature wird mit origin/feature verknüpft
git push -u origin feature

# funktioniert nur mit -u automatisch
git push / git pull

# Verbinde deinen Branch explizit mit dem Remote-Branch:
git branch --set-upstream-to=origin/main main


git remote show origin
```
