#### Lokales Repo auf Github bringen

Wenn wir unser lokales Repo *GitDemo* auf Github bringen wollen, legen wir zunächst auf Github ein Repo *GitDemo* an und kopieren die html-Adresse dieses Repos.

<img src='github01.png' width='901'>

```python
git remote add origin https://github.com/ktheu/GitDemo.git
git push origin main
```

Der main-branch des lokalen Repos wird nach nach Github gepushed. *origin* ist der Name für die html-Adresse, es ist der Standardname für das remote-Repo. <br>
Wenn wir immer wieder Änderungen vom selben branch ins remote-Repo pushen wollen, dann können wir einmal eingeben:

```python
git push -u origin main 
```

Dann reicht (sofern wir auf dem main-branch sind) das einfache

```python
git push
```

Das geht auch mit anderen branches. Hier wird der lokale Feature branch mit dem feature branch auf Github verknüpft.
Wenn man danach im feature branch ist, reicht das einfache *git push*

```python
git push -u origin feature
```

```python
git push --all origin      # pushed alle branches nach Github
```

#### Mit neuem Repo lokal und auf Github starten

Wenn wir ein neues Repo lokal und auf Github anlegen wollen, legen wir es zunächst auf Github an und clonen dies dann auf unseren lokalen Rechner.
Beim Clone wird der Ordner mit dem lokalen Repo angelegt. Den clone Befehl führen wir also aus in dem Ordner *in den hinein* das lokale Repo erstellt werden soll.

```python
git clone https://github.com/ktheu/GitDemo2.git
```

#### Collaborators

Wir können zu unserem Repo Mitarbeiter hinzufügen (Collaborators). Wir verschicken dazu eine Einladung, die der Adressat annehmen oder ablehnen kann. Der Collaborator hat immer Schreibrechte, d.h. er kann Code pushen, mergen, pull-requests akzeptieren usw. aber keine Admin-Rechte, also er kann z.B. keine neuen Mitarbeiter aufnehmen.


#### pull-requests

Mit pull-requests können wir Änderungen in einem branch vorschlagen. Pull-requests können diskutiert, geändert, verworfen oder akzeptiert werden.

#### Zusammenarbeit mit Anderen

Wir betrachten zwei Workflows:

- Workflow 1: Feature Branches for Collaborators
- Workflow 2: Fork & Clone for Contributors 

Beispiel Workflow 1: Zwei Leute arbeiten als Collaborators an einem gemeinsamen Programmierprojekt. Jeder arbeitet an einem anderen Feature in einem zu dem Feature gehörenden branch. Der main-branch wird nur upgedated bei allgemein akzeptierten Änderungen. Diskussionen und Akzeptanz von Änderungen finden mittels pull-requests statt.

<img src='github08.png' width='800'>

Beispiel Workflow 2: Wir wollen (als Contributor) einen Beitrag zu einem Open-source Programmierprojekt beisteuern. Wie haben nicht die Rolle eines Collaborators. Dazu forken wir das Zielrepo in Github und clonen den fork lokal. Unsere Änderungen schlagen wir als pull-request im Original-Repo vor.
