### Github

Wir nutzen Github

- als Cloudspeicher für unsere lokalen Repos
- um von unterscheidlichen PCs (Home-Schule) das gleiche Programmierprojekt zu bearbeiten
- zur Zusammenarbeit mit Anderen

#### Lokales Repo auf Github bringen

Wenn wir unser lokales Repo *GitDemo* auf Github bringen wollen, legen wir zunächst auf Github ein Repo *GitDemo* an und kopieren die html-Adresse dieses Repos.

```python
https://github.com/ktheu/GitDemo.git
```

In unserem lokalen Repo geben wir dann ein:

```python
git remote add origin https://github.com/ktheu/GitDemo.git
```

*origin* ist jetzt der Name für die html-Adresse, es ist der Standardname für das remote-Repo.

Dann geben wir ein:

```python
git push origin main
```

Dadurch wird der main-branch des lokalen Repos nach Github gepostet. <br>
Wenn wir immer wieder Änderungen vom selben branch ins remote-Repo pushen wollen, dann können wir einmal eingeben:

```python
git push -u origin main 
```

Dann reicht anschließend das einfache

```python
git push
```

Das geht auch mit anderen branches. Hier wird der lokale Feature branch mit dem feature branch auf Github verknüpft.
Wenn man danach im feature branch ist, reicht das einfache *git push*

```python
git push -u origin feature
```

```python
git push --all origin      # pushed alle branches nach Github
```

#### Neues Repo lokal und auf Github anlegen

Wenn wir ein neues Repo lokal und auf Github anlegen wollen, legen wir es zunächst auf Github an und clonen dies dann auf unseren lokalen Rechner.
Beim Clone wird der Ordner mit dem lokalen Repo angelegt. Den clone Befehl führen wir also aus in dem Ordner *in den hinein* das lokale Repo erstellt werden soll.

```python
git clone https://github.com/ktheu/GitDemo2.git
```

#### Von unterschiedlichen PCs das gleiche Programmierprojekt bearbeiten

Wir melden uns am PC1 und PC2 mit demselben Github-Account an.
Wenn wir am PC1 eine Änderung nach Github gepushed haben, dann holen wir uns an PC2 mit pull diese Änderung ab, bevor wir dort weiterarbeiten.
