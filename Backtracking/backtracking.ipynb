### Backtracking

Backtracking ist ein systematisches Ausprobieren. Wenn wir in eine Sackgasse geraten sind, gehen wir zurück (backtrack) und probieren die nächste Variante aus. Meistens sind alle möglichen Lösungen gefragt.
Backtracking ist anwendbar, wenn wir eine Lösung schrittweise aufbauen können und wenn wir für eine Teillösung die möglichen Kandidaten aufzählen können, die die Teillösung erweitern.
Beim Backtracking wird eine Teillösung immer wieder verändert. Erst wenn eine Lösung gefunden wird, wird eine Kopie gesichert.
Eine Teillösung bezeichnen wir mit a. a ist eine Liste mit Entscheidungen, die zur Lösung führen sollen.

Die Funktionen im Muster kann man im konkreten Fall kompakter in die back-Funktion integrieren.

##### Muster für das Backtracking

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
        return                       
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            # ggf weitere Operationen
            back(a)
            # ggf weitere Operationen rückgängig machen
            a.pop()
                
def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    pass

def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern können
    '''
    pass

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    pass

solutions = []
back([])
print(solutions)
```

#### Beispiel 2:  8-Damen Problem

Wieviele Möglichkeiten gibt es, 8 Damen auf einem Schachbrett so zu positionieren, dass sie sich gegenseitig nicht bedrohen?
- die Frage wurde 1848 in der Berliner Schachzeitung veröffentlicht.
- 1850 wurde erstmals die richtige Antwort gegeben:  92

Eine Teillösung repräsentieren wir wieder durch eine Liste

    a = [0, 6, 3, 5, 7]

steht für die folgende Teillösung:

    In Reihe mit Index 0 steht eine Dame auf Spalte mit Index 0
    In Reihe mit Index 1 steht eine Dame auf Spalte mit Index 6
    In Reihe mit Index 2 steht eine Dame auf Spalte mit Index 3
    In Reihe mit Index 3 steht eine Dame auf Spalte mit Index 5
    In Reihe mit Index 4 steht eine Dame auf Spalte mit Index 7

<img src='damen1.png' width='300'>

Kandidaten für eine Erweiterung der Teillösung sind die Zahlen 0 - 7, aber nur die Zahlen 1 und 4 sind gute Kandidaten.

<img src='damen2.png' width='601'>

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
        return
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            back(a)
            a.pop()
                
def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    return len(a) == anzahl_damen

def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern
    '''
    return range(anzahl_damen)

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    for j in range(len(a)):
        if a[j] == cand: return False
        if len(a) - j == abs(cand-a[j]): return False
    return True

def printLoesung(a):
    for i in range(anzahl_damen):
        for j in range(anzahl_damen):
            print('X' if a[i]==j else '.', end=' ')
        print()
    print()

anzahl_damen = 4
solutions = []
back([])
print('Anzahl Lösungen:',len(solutions))
for k, a in enumerate(solutions):
    print(f'Lösung {k+1}:')
    printLoesung(a)
```

#### Beispiel 3:  Alle Kreise in einem Graphen

<img src='kreise1.png' width='300'>

Mit backtracking ermitteln wir die einfachen Kreise, die von einem gegebenen Knoten ausgehen.

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
        return
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            back(a)
            a.pop()
                
def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    return len(a) > 1 and a[-1] == a[0]

def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern
    '''
    return G[a[-1]]

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    return cand not in a[1:] 

G = {'a':{'b'}, 'b':{'c'}, 'c':{'d','f'}, 'd':{'e'}, 'e':{'f'}, 'f':{'a','c','e'}}
solutions = []
back(['f'])
print(solutions)
```

Kompaktere Fassung:

```python
def kreise(G, v):
    '''
    G Graph, v Knoten des Graphen
    returns: Liste aller einfachen Kreise von v nach v.
    '''
    def back(a):
        if len(a) > 1 and a[-1] == a[0]:
            solutions.append(a.copy())
            return
        for cand in  G[a[-1]]:
            if cand not in a[1:]:
                a.append(cand)
                back(a)
                a.pop()
    solutions = []
    back([v])
    return solutions


G = {'a':{'b'}, 'b':{'c'}, 'c':{'d','f'}, 'd':{'e'}, 'e':{'f'}, 'f':{'a','c','e'}}
kreise(G,'f')
 
```

Damit Kreise nicht doppelt gezählt werden, wenn wir von verschiedenen Knoten aus starten, repräsentieren wir sie mit ihrer sortierten Kantenliste.

```python
def getKanten(kreis):
    tmp = []
    for i in range(len(kreis)-1):
        tmp.append(kreis[i]+kreis[i+1])
    tmp.sort()
    return tmp

getKanten(['f', 'c', 'd', 'e', 'f'])
```

##### Alle Kreise eines Graphen

```python
def alleKreise(G):
    tmp = []
    for v in G:
        for kreis in kreise(G,v):
            kanten = getKanten(kreis)
            if kanten not in tmp:
                tmp.append(kanten)
    return tmp

alleKreise(G)
```
