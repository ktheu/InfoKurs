### Backtracking

[Video](https://youtu.be/j4tds6gD3fQ)  - [Aufgabenvidoe](https://youtu.be/H_9N8_OGHso)

Backtracking wenden wir an, wenn alle möglichen Lösungen für ein Problem gefragt sind. Wir erweitern schrittweise Teillösungen zu einer Gesamtlösung. Wenn wir in eine Sackgasse geraten sind oder eine Lösung gefunden haben, nehmen wir unsere letzten Entscheidungen zurück und probieren weitere Varianten aus. 

Beispiel: Wir wollen alle Permutationen von [1,2,3] finden.

Meistens sind alle möglichen Lösungen gefragt.
Backtracking ist anwendbar, wenn wir eine Lösung schrittweise aufbauen können und wenn wir für eine Teillösung die möglichen Kandidaten aufzählen können, die die Teillösung erweitern.
Beim Backtracking wird eine Teillösung immer wieder verändert. Erst wenn eine Lösung gefunden wird, wird eine Kopie gesichert.
Eine Teillösung ist eine Liste mit Entscheidungen, die zur Lösung führen sollen.

Die Funktionen im Muster kann man im konkreten Fall kompakter in die back-Funktion integrieren.

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
        return                       
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            # ggf weitere Operationen
            back(a)
            # ggf weitere Operationen rückgängig machen
            a.pop()
                
def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    pass

def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern können
    '''
    pass

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    pass

# Aufruf
# solutions = []
# back([])
# print(solutions)
```

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
        return                       
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            back(a)
            a.pop()

def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    return len(a) == n

def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern können
    '''
    return list(range(1,n+1))

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    return cand not in a

n = 3
solutions = []
back([])
print(solutions)
```

#### 8-Damen Problem

Wieviele Möglichkeiten gibt es, 8 Damen auf einem Schachbrett so zu positionieren, dass sie sich gegenseitig nicht bedrohen?
- die Frage wurde 1848 in der Berliner Schachzeitung veröffentlicht.
- 1850 wurde erstmals die richtige Antwort gegeben:  92

Wir überlegen uns das Vorgehen an dem kleinerem 4-Damen-Problem. Wir können uns eine mögliche Lösung schrittweise aufbauen, indem wir nacheinander entscheiden, in welche Spalte wir die Damen setzen. Wir beginnen mit der Dame ganz oben (Zeile 0) und gehen dann schrittweise nach unten.

<img src='damen4.png'>

Wir müssen insgesamt 4 Entscheidungen treffen. Wenn wir eine Teillösung gefunden haben, dann können wir von allen Kandidaten 0,1,2,3 herausfinden, ob sie die Teillösung sinnvoll erweitern, indem wir auf mögliche Kollisionen prüfen.

<img src='damen5.png' width='600'>

Um zu beurteilen, ob ein Kandidat die Teillösung sinnvoll erweitert, prüfen wir, ob die neu zu setzende Dame eine schon gesetzte Dame bedroht.

<img src='damen6.png'>

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
        return                       
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            back(a)
            a.pop()

def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    return len(a) == anzahl_damen

def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern können
    '''
    return list(range(anzahl_damen))

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    for i in range(len(a)):
        if a[i] == cand: return False
        if len(a) - i == abs(cand-a[i]): return False
    return True

def printLoesung(a):
    for i in range(anzahl_damen):
        for j in range(anzahl_damen):
            print('X' if a[i]==j else '.', end=' ')
        print()
    print()

anzahl_damen = 4
solutions = []
back([])
print(solutions)
print('Anzahl Lösungen:',len(solutions))
for k, a in enumerate(solutions):
    print(f'Lösung {k+1}:')
    printLoesung(a)
```
