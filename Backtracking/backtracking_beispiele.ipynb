#### Jugendherberge

In der Jugendherberge organisiert man jeden Abend ein Lagerfeuer. Am ersten Abend setzen
sich alle Jugendlichen rund ums Feuer. Am zweiten Abend nehmen sie so Platz, dass die, 
die am ersten Abend Nachbarn waren, diesmal nicht nebeneinander sitzen. 
Bestimme alle Möglichkeiten, wie sie sich am zweiten Abend setzen können. Wir nehmen an,
dass sie am ersten Abend mit 1,2,3,...,n nummeriert sind.

<img src='lagerfeuer.png' width='800'>

Ein Kandidat erweitert eine Teillösung sinnvoll, wenn der Abstand zum letzten Element der Teillösung weder 1 noch n-1 ist.

<img src='lagerfeuer1.png' width='300'>

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
        return                       
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            back(a)
            a.pop()
                
def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    return len(a) == n
 
def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern können
    '''
    return list(range(1,n+1))
 
def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    if cand in a: return False
    if len(a) > 0 and (abs(a[-1]-cand) == 1 or abs(a[-1]-cand) == n-1): return False
    return True
```

```python
n = 5
solutions = []
back([])
print(f'Es gibt {len(solutions)} Möglichkeiten:')
for x in solutions:
    print(*x)
```

#### Referate
Die Erdkunde-Lehrerin möchte, dass die Schülerinnen und Schüler Referate halten, und verteilt die Themen.
Sie hat für die n Schülerinnen und Schüler m Themen vorbereitet (m <= n), von denen jedes mindestens einmal bearbeitet werden soll.
Welche Möglichkeiten gibt es, die Themen zu verteilen?

<img src='referate.png'>
 

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
        return
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            back(a)
            a.pop()
                
def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    return len(a) == n  

def candidates(a):  
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern können
    '''
    return list(range(1,m+1))

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    k = len(set(a) | {cand} )             # Anzahl schon vergebener verschiedenen Themen inclusive des aktuellen Kandidaten
    if m-k > n-len(a)-1: return False     # Anzahl noch zu vergebender Themen ist größer als die Anzahl der noch nicht vergebenen Schüler
    return True

           
n = 3
m = 2
solutions = []
back([])
print(f'Es gibt {len(solutions)} Möglichkeiten:')
for x in solutions:
    print(*x)
```

#### Haus vom Nikolaus

Wieviele Möglichkeiten gibt es, das Haus vom Nikolaus beginnend von der linken unteren Ecke zu zeichnen?
Das Haus hat 5 Knoten und 8 Kanten. Wir beginnen immer mit Knoten 0. Dann treffen wir 8 Entscheidungen (welches ist unser nächster Knoten?) Wir können schrittweise verfolgen, ob eine Entscheidung eine Teillösung sinnvoll erweitert. Dazu merken wir uns, welche Kanten wir bisher schon verbraucht haben.

<img src='nikolaus.png' width='600'>

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
        return
    for cand in candidates(a):
        if isGood(cand,a):
            used[(a[-1], cand)] = True    # diese Kanten sind verbraucht
            used[(cand, a[-1])] = True
            a.append(cand)
            back(a)
            a.pop()
            used[(a[-1], cand)] = False   # diese Kanten sind wieder unbenutzt
            used[(cand, a[-1])] = False
                
def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    return len(a) == 9

def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern
    '''
    return list(range(5))

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    if (a[-1], cand) not in edges: return False
    return not used[(a[-1], cand)]
```

```python
edges = [(0,1),(1,0),(1,2),(2,1),(2,3),(3,2),(3,4),(4,3),(4,0),(0,4),(1,3),(3,1),(0,3),(3,0),(1,4),(4,1)]
used = {e: False for e in edges}

solutions = []
back([0])
print(len(solutions))
for k, sol in enumerate(solutions):
    print(f'{k+1:2}:', *sol)
```

#### Sudoku

Das Vorgehen überlegen machen wir uns an einem 4 x 4 Sudoku klar. Ein ordentliches Sudoku hat nur eine Lösung. Für den Fall, dass es doch mehrere Lösungen gibt, soll unser Programm alle Lösungen finden. 

<img src='sudoku1.png' width='500'>

Bei diesem Sudoku haben wir 11 Entscheidungen zu treffen. Als Kandidaten stehen uns die Zahlen von 1 bis 4 zur Verfügung. Für jeden Kandidaten können wir entscheiden, ob er die Teillösung sinnvoll erweitert.

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
        return
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            grid[leer[len(a)-1]] = cand
            back(a)
            grid[leer[len(a)-1]] = 0
            a.pop()
            
                
def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    return len(a) == len(leer)

def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern
    '''
    return list(range(1,n+1))

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    x, y = leer[len(a)]     # die nächste zu füllende Leerstelle
    x1 = x - x%sn           # die linke obere Ecke  
    y1 = y - y%sn           # des zugehörigen Quadrats
    quadrat = grid[x1:x1+sn,y1:y1+sn].flatten()    
    if cand not in grid[x,:] and cand not in grid[:,y] and cand not in quadrat:
        return True
    return False
```

```python
data = '''
.2..
1.4.
...3
3...
'''
```

```python
zeilen = data.strip().split()
zeilen
```

```python
import numpy as np
n = len(zeilen)
sn = int(np.sqrt(n))
n, sn
```

```python
grid = np.zeros((n,n),dtype='int')
grid
```

```python
for x in range(n):
    for y in range(n):
        grid[x,y] = int(zeilen[x][y]) if zeilen[x][y].isdigit() else 0
grid
```

In der Liste leer speichern wir die Koordinaten der Leerfelder

```python
leer = []
for x in range(n):
    for y in range(n):
        if grid[x][y] == 0:
            leer.append((x,y))
print(leer)
```

Eine mögliche Lösung ist eine Liste der gleichen Länge wie leer, die jeder Leerstelle einen der möglichen Kandidaten 1,2,3,4 zuweist.

Um zu prüfen, ob ein Kandidat für die Leerstelle (x,y) gut ist, müssen wir auch das Quadrat überprüfen, in dem die Leerstelle liegt.

<img src='sudoku2.png' width='800'>

```python
solutions = []
back([])
print(f'Anzahl der Lösungen: {len(solutions)}')

for i, solution in enumerate(solutions):
    grid0 = grid.copy()
    print(f'Lösung {i+1}:')
    for k in range(len(leer)):
        x, y = leer[k]
        grid0[x, y] = solution[k]
    print(grid0)
```

#### Alle Pfade

In einem DAG (directed acyclic graph) wollen wir alle Pfade von einem Startknoten zu einem Zielknoten bestimmen.
Wir fassen den Code gleich zusammen. Ob die Nachbarn eines Knoten als Menge oder als Liste (ohne Wiederholung) gegeben ist, spielt keine Rolle.

<img src='graph2.png' width='301'>

```python
def allPaths(G, start, end):
    '''
    returns: Liste aller Pfade von start nach end in einem DAG (directed acyclic graph) G als Listen.

    G = {0:[4,3,1], 1:[3,2,4], 2:[3], 3:[4], 4:[]}
    >>> allPaths(G, 0, 4)
    [[0, 4], [0, 3, 4], [0, 1, 3, 4], [0, 1, 2, 3, 4], [0, 1, 4]]
    '''
    def back(a):
        if a[-1] == end:
             solutions.append(a.copy())
        for cand in G[a[-1]]:
            a.append(cand)
            back(a)
            a.pop()
                    
    solutions = []
    back([start])
    return solutions
```

```python
G = {0:[4,3,1], 1:[3,2,4], 2:[3], 3:[4], 4:[]}
allPaths(G, 0, 4)
```
