### Beamsearch

Falls A* zu langsam ist, können wir versuchen mit Beamsearch eine einigermaßen gute Lösung zu finden. Wir geben uns eine Zahl *beamwidth* vor. Nachdem die Breitensuche eine Ebene untersucht hat, werden in der frontier nur die best *beamwidth* states behalten und weiter untersucht.
