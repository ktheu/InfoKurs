### Search: Breitensuche (bfs)

In vielen Problemstellung geht es darum, in einer (meist sehr großen) Menge von Möglichkeiten eine Lösung zu finden. Am Beispiel des 8-Puzzle werden verschiedene Suchalgorithmen für das Finden einer Lösung vorgestellt.

Einzelne Spielstellungen stellen wir uns als Knoten vor. Die Kanten zwischen den Knoten sind mögliche Spielzüge.
Die Startstellung ist also die Wurzel eines Suchbaums, in dem wir einen Pfad zu einem Knoten suchen, der den **goaltest** besteht. 

In der Regel ist der Suchbaum so groß, dass er nicht in seiner Gesamtheit aufgebaut werden kann. Während der Suche wird der Suchbaums immer wieder erweitert und wir hoffen, dass wir auf eine Lösung stoßen, ohne den gesamten Baum durchsuchen zu müssen.

Die Knoten, die wir noch untersuchen müssen, verwalten wir in der **frontier** (fringe, open set). 
Zu jedem Knoten, der in die frontier kommt, merken wir uns den Elternknoten in einem dictionary **prev**.
Die keys in prev sind also alle Spielstellungen, die schon untersucht wurden oder noch auf Untersuchung warten.

Zu Beginn ist nur der startstate in der frontier. Der startstate hat den Elternknoten *None*.

Wenn wir einen state untersuchen, holen wir ihn aus der frontier und schauen, ob er den goaltest besteht.

Wenn das nicht der Fall ist, gehen wir die Liste seiner Kinder durch. Nur Kinder, die noch nicht untersucht wurden oder die noch nicht auf Untersuchung warten, werden der frontier hinzugefügt. 

Wenn wir einen state gefunden haben, der den goaltest besteht, beenden wir die Suche und geben den gefunden state zusammen mit dem dictionary **prev** zurück. Aus beiden berechnet uns die Funktion **reconstructPath** den Weg vom Start- bis zum Zielzustand.

Die Funktion **getMove** beschreibt den Übergang von einem state zum nächsten mit sinnvollen Worten.

Für jeden Anwendungsfall müssen wir die Funktionen **nextstates, goaltest, getMove** implementieren.

Für den Fall des 8-Puzzle:

```python
def goaltest(state):
    return state == (0,1,2,3,4,5,6,7,8) 

def nextstates(state):
    tmp = []
    i = state.index(0)
    for d in [-3,3,-1,1]:
        j = i+d
        if 0 <= j < 9 and (i%3==j%3 or i//3==j//3):
            a = list(state)
            a[i],a[j] = a[j],a[i]
            tmp.append(tuple(a))
    return tmp

def getMove(s1, s2):
    i1 = s1.index(0)   # Position der 0 in state s1  
    i2 = s2.index(0)   # Position der 0 in state s2    
    if i2 == i1-1: return 'L'   #left
    if i2 == i1+1: return 'R'   #right
    if i2 == i1+3: return 'D'   #down
    if i2 == i1-3: return 'U'   #up
```

Um eine Instanz des Online-Spiels zu lösen, müssen wir den goalstate auf (1,2,3,4,5,6,7,8,0) ändern und uns nach dem shuffle, die Nummern anzeigen lassen.

```python
def bfs2(startstate):
    frontier =  deque([startstate])
    prev = {startstate:None}
    explored = []
    while frontier:
        state = frontier.popleft()  
        explored.append(state)
        if goaltest(state):
            return prev, state
        for v in nextstates(state):
            if v not in frontier and v not in explored:
                frontier.append(v)
                prev[v] = state
    return None, None
```

#### Heuristik
Wir wollen aus der frontier die Spielstellungen früh entnehmen, die Erfolg versprechen. Dazu bewerten wir sie mit einer Heuristik. 
Als Datenstruktur eignet sich ein Heap.


**1. Heuristik:** Die Anzahl der Ziffern, die ihre Endposition noch nicht erreicht haben, betrachten wir als die vermuteten Vorwärtskosten. 

```python
def h(state):
    '''
    state: Spielstellung
    returns: Vorwärtskosten laut Heuristik
    '''
    return sum(a!=b for a,b in zip((0,1,2,3,4,5,6,7,8),state))
```

### Search: Greedy
Der Greedy-Algorithmus nimmt die Spielstellung mit den niedrigsten Vorwärtskosten als nächstes dran. 

```python
from heapq import heappop, heappush
def greedy(startstate):
    frontier =[(h(startstate),startstate)]  
    prev = {startstate:None}
    while frontier:
        _ ,state = heappop(frontier)  
        if goaltest(state):
            return prev,state
        for v in nextstates(state):
            if v not in prev:
                heappush(frontier,(h(v),v))
                prev[v] = state
    return None, None
```

```python
%%time
prev, state = greedy(startstate)
if prev is None:
    print('Keine Lösung gefunden')
else:
    path = reconstructPath(prev, state)
    print(f'Lösung in {len(path)-1} Schritten')
    for i in range(len(path)-1):
        print(getMove(path[i], path[i+1]),end=' ')
    print()
```

**Greedy-Algorithmen sind meist schnell, finden aber nicht immer die optimale Lösung.**


### Search: A*

Der A*-Algorithmus berücksichtigt sowohl die vermuteten Vorwärtskosten laut Heuristik, als auch die bisher
tatsächlich angefallenen (Rückwärts)-Kosten.

```python
from heapq import heappop, heappush
def astar(startstate):
    frontier =[(h(startstate),startstate)]  
    prev = {startstate:None}
    g = {startstate:0}          # bisher angefallenen Kosten: die Anzahl Züge
    while frontier:
        _ ,state = heappop(frontier)   
        if goaltest(state):
            return prev,state
        for v in nextstates(state):
            gg = g[state] + 1
            if v not in prev or gg < g[v]:   # v noch nicht explored und nicht in frontier oder die Rückwärtskosten sind geringer
                g[v] = gg                    # als die bisherigen Rückwärtskosten von v.
                heappush(frontier,(g[v]+h(v),v))
                prev[v] = state
    return None, None
```

```python
%%time
prev, state = astar(startstate)
if prev is None:
    print('Keine Lösung gefunden')
else:
    path = reconstructPath(prev, state)
    print(f'Lösung in {len(path)-1} Schritten')
    for i in range(len(path)-1):
        print(getMove(path[i], path[i+1]),end=' ')
    print()
```

**2.Heuristik:** Die Manhatten-Distanz der Ziffern zu ihrer Zielposition

```python
def h(state): 
    '''
    state: Spielstellung
    returns: Fortwärtskosten laut Manhatten-Heuristik
    '''
    mh = 0
    for i in range(1,9):
        if state[i] != 0:
            z1,s1 = i//3, i%3
            z2,s2 = state[i]//3, state[i]%3
            mh += abs(z1-z2)+abs(s1-s2)
    return mh
```

```python
%%time
prev, state = astar(startstate)
if prev is None:
    print('Keine Lösung gefunden')
else:
    path = reconstructPath(prev, state)
    print(f'Lösung in {len(path)-1} Schritten')
    for i in range(len(path)-1):
        print(getMove(path[i], path[i+1]),end=' ')
    print()
```



Eine Heuristik heißt **zulässig**, wenn sie die tatsächlich anfallenden Kosten nicht überschätzt. Mit einer zulässigen Heuristik findet A* die optimale Lösung.
