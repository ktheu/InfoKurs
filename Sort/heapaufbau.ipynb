### Aufbau eines Heaps in O(n)

[Video](https://youtu.be/nI0itPEURfI)

<img src='heapaufbau1.png' width='600'>

```python
summe = 0
for k in range(1,100):
    summe += k/(2**(k+1))
print(summe)
```

<img src='heapaufbau2.png' width='800'>

```python
x = 0.25+0.25j
RS = 1/(1-x)

summe = 0
for k in range(1000):
    summe += x**k
print(summe,RS)
```
