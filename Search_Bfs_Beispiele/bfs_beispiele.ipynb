### Search mit BFS - Beispiele

[Video1](https://youtu.be/xRZvhyEfvr4) - [Video2](https://youtu.be/7LW-xa43GZc) - [Video3](https://youtu.be/l5STsdlJ7X4)


Die untenstehenden drei Funktionen können wir unverändert übernehmen. In jedem Anwendungsfall müssen wir 

    - einen state modellieren (meist Tuple oder String).
    - die Funktionen goaltest und nextstates implementieren
    - ggf. die Funktion getMove implementieren.


````python
from collections import deque
def bfs(startstate):
    frontier =  deque([startstate])
    prev = {startstate:None}
    while frontier:
        state = frontier.popleft()  
        if goaltest(state):
            return prev, state
        for v in nextstates(state):
            if v not in prev:
                frontier.append(v)
                prev[v] = state
    return None, None

def reconstructPath(prev,goalstate):
    state = goalstate
    path = []
    while state is not None:
        path.append(state)
        state = prev[state]
    path.reverse()
    return path

def getMoves(path):
    moves = ''
    for i in range(len(path)-1):
        moves+=getMove(path[i],path[i+1])
    return moves
````

#### MOO-Operations

Gegeben ist ein String s der nur aus den Zeichen M und O besteht. Folgende Operationen sind möglich:
- (1) ersetze den ersten oder letzten Buchstaben durch den jeweils anderen ('M' wird zu 'O' bzw. 'O' wird zu 'M')
- (2) lösche den ersten oder letzten Buchstaben

Ziel ist es, durch eine minimale Anzahl von Operationen aus dem Anfangsstring s den String 'MOO' zu erzeugen.
Ermittle die Anzahl der Operationen und beschreibe einen Weg dorthin. 

Hinweis: BFS ist hier nicht die schnellste Methode für die Lösung, aber zu Übungszwecken eignet sich die Aufgabe.

    Beispiel:
    s = 'MOMMOM'

    Lösung in 4 Schritten:
    ändere letzten Buchstaben: MOMMOO
    erster Buchstabe weg: OMMOO
    erster Buchstabe weg: MMOO
    erster Buchstabe weg: MOO

Wir nutzen den String als state

<img src='moo1.png' width='701'>

````python
def nextstates(state):
    tmp = []
    if state[0] == 'M':
        tmp.append('O'+state[1:])
    else:
        tmp.append('M'+state[1:])
    if state[-1] == 'O':
        tmp.append(state[:-1]+'M')
    else:
        tmp.append(state[:-1]+'O')
    tmp.append(state[1:])
    tmp.append(state[:-1])
    return tmp  

def goaltest(state):
    return state == 'MOO'

def getMove(s1, s2):
    if len(s1)>len(s2):
        if s1[1:]==s2:
            return 'erster Buchstabe weg\n'
        return 'letzter Buchstabe weg\n'
    else:
        if s1[0]==s2[0]:
            return 'ändere letzten Buchstaben\n'
        return 'ändere ersten Buchtaben\n'
````

````python
#start = 'MOMMOM'
start = 'MMOMOMOMM'
prev, goal = bfs(start)
path = reconstructPath(prev,goal)
print(path)
print(getMoves(path))
````

#### Girlande

Eine Girlande besteht aus vier Lampen, die vier verschiedene Farben haben können. Ein String s der Länge 4, der nur die Zeichen 1 bis 4 enthält, stellt eine Girlande dar.

```
# Beispiel:
girlande = '1211'     # Girlande besteht aus 1 Lampe mit Farbe 1, 1 Lampe mit Farbe 2, dann wieder 2 Lampen mit Farbe 1
```

Zu Beginn sind alle Lampen ausgeschaltet. Wir können beliebig oft folgende Operation durchführen:
Wähle eine Lampe und ändere ihren Status (aus 'An' mache 'Aus' und umgekehrt). Die nächste Operation darf aber nur an einer Lampe ausgeführt werden, die eine andere Farbe hat. 

Berechne die minimale Anzahl der Operationen, die wir benötigen, um alle Lampen der Girlande anzuschalten und beschreibe eine mögliche Lösung.
Gib 'Keine Lösung' aus, wenn es nicht geht.

```
# Beispiel:
girlande='1211'
Anzahl Operationen: 6
Lampe 1 ein
Lampe 2 ein
Lampe 3 ein
Lampe 2 aus
Lampe 4 ein
Lampe 2 ein


Wie sollen wir einen state modellieren?

<img src='girlande1.png' width='700'>

Um aus einem state die nextstates zu ermitteln, benötigen wir noch die Information, welche Farbe als letztes gewechselt wurde.

<img src='girlande2.png' width='700'>

````python
def nextstates(state):
    tmp = []
    tup, letzte_farbe = state   # k letzte benutzte Farbe
    for i in range(4):
        farbe = abs(tup[i])
        if farbe != letzte_farbe:
            b = list(tup)
            b[i] = -b[i]
            tmp.append((tuple(b),farbe))
    return tmp

def goaltest(state):
    return state[0] == ziel

def getMove(s1, s2):
    t1 = s1[0]
    t2 = s2[0]
    for i in range(len(t1)):
        if t1[i] != t2[i]:
            break
    if t2[i] < 0:
        return f'Lampe {i+1} aus\n'
    else:
        return f'Lampe {i+1} ein\n'

#girlande = '1211'
#girlande = '1111'
girlande = '1231'
start = tuple([-int(x) for x in girlande])   # zu Beginn sind alle Lampen ausgeschaltet
ziel = tuple([int(x) for x in girlande])     # am Ende sind alle Lampen eingeschaltet
startstate = (start,0)                         # zu Beginn wurde vorher keine Farbe benutzt
prev, state = bfs(startstate)
print(f'{girlande=}')
if prev is None:
    print('Keine Lösung')
else:
    path = reconstructPath(prev, state)
    print(f'Anzahl Operationen: {len(path)-1}')
    print(getMoves(path))
    
````

#### 8Puzzle

<img src='such_bild1.png' width='500'>

Wir modellieren einen state als Tuple mit der Reihenfolge der Zahlen, die Leerstelle bekommt die Null. Die Aktionen beschreiben wir als Bewegung der Leerstelle.

<img src='8puzzle.png' width='900'>

````python
def goaltest(state):
    return state == (0,1,2,3,4,5,6,7,8) 

def nextstates(state):
    tmp = []
    i = state.index(0)
    for d in [-3,3,-1,1]:
        j = i+d
        if 0 <= j < 9 and (i%3==j%3 or i//3==j//3):
            a = list(state)
            a[i],a[j] = a[j],a[i]
            tmp.append(tuple(a))
    return tmp

def getMove(s1, s2):
    i1 = s1.index(0)   # Position der 0 in state s1  
    i2 = s2.index(0)   # Position der 0 in state s2    
    if i2 == i1-1: return 'L'   #left
    if i2 == i1+1: return 'R'   #right
    if i2 == i1+3: return 'D'   #down
    if i2 == i1-3: return 'U'   #up
````

````python
startstate = (7,2,4,5,0,6,8,3,1)   
prev, goal = bfs(startstate)
path = reconstructPath(prev,goal)
print(f'Anzahl Operationen: {len(path)-1}')
print(getMoves(path))
````
