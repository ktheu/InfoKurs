### Uniform_Cost Search

Uniform_cost erweitert die Suche (möglichst) gleichmäßig gemessen an den Kosten. Wenn man keine spezielles Ziel sucht sondern alle erreichbaren Knoten besucht, nennt man diesen Algorithmus: Dijkstra's Algorithmus.

```python
from heapq import heappop, heappush
def uniform_cost(s):
    '''
    s: startstate
    returns: prev, goalstate, dist falls erfolgreich
       dist[goalstate] ist die Länge des kürzesten Wegs von s zu goalstate
    '''
    frontier =[(0,s)]  
    prev = {s:None}
    dist = {s:0}                
    while frontier:
        _ ,state = heappop(frontier) 
        if goaltest(state):
            return prev, state, dist
        for v in nextstates(state):
            if v not in prev or dist[state] + cost(state,v) < dist[v]:
                dist[v] = dist[state] + cost(state,v)
                heappush(frontier,(dist[v],v))
                prev[v] = state
    return None, None, None

```
