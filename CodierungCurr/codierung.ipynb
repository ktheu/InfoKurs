### ASCII, ANSI, UTF-8



| Merkmal                                    | ANSI (z. B. Windows-1252)                    | UTF-8                                                     |
| ------------------------------------------ | -------------------------------------------- | --------------------------------------------------------- |
| **Zeichenvorrat**                          | Maximal 256 Zeichen                          | Alle Unicode-Zeichen (über 1 Million mögliche Codepunkte) |
| **Speicherbedarf für ASCII-Zeichen**       | 1 Byte                                       | 1 Byte                                                    |
| **Speicherbedarf für Umlaute (ä, ö, ü)**   | 1 Byte                                       | 2 Byte                                                    |
| **Speicherbedarf für chinesische Zeichen** | Nicht darstellbar                            | Meist 3 Byte                                              |
| **Speicherbedarf für Emojis**              | Nicht darstellbar                            | Meist 4 Byte                                              |
| **Variable Zeichenlänge**                  | Nein (immer 1 Byte)                          | Ja (1–4 Byte)                                             |
| **Kompatibilität zu ASCII**                | Die ersten 128 Zeichen entsprechen ASCII     | Vollständig ASCII-kompatibel                              |
| **Sprachunterstützung**                    | Meist nur eine Sprachregion gleichzeitig     | Weltweit einsetzbar                                       |
| **Austausch zwischen Systemen**            | Problematisch (verschiedene ANSI-Codepages)  | Standard im Internet                                      |
| **Heutige Bedeutung**                      | Veraltet, vor allem ältere Windows-Programme | Aktueller Standard                                        |


ANSI

Einfach aufgebaut.
Jede Region benötigt oft eine andere Codepage (z. B. Deutsch, Russisch, Griechisch).
Dateien können auf anderen Systemen falsch dargestellt werden.

UTF-8

Einheitliche Kodierung für alle Sprachen.
Verhindert viele Darstellungsprobleme.
Standard für Webseiten, E-Mails und moderne Betriebssysteme.
Rückwärtskompatibel zu ASCII.

ANSI ist bei einigen europäischen Texten etwas platzsparender, unterstützt aber nur einen sehr begrenzten Zeichenvorrat. UTF-8 benötigt für viele Sonderzeichen mehr Speicher, kann dafür jedoch alle Unicode-Zeichen darstellen und hat sich deshalb als weltweiter Standard durchgesetzt.

# Bedeutung der Datenkomprimierung im Alltag

## Vorteile der Datenkomprimierung

### 1. Weniger Speicherplatz
- Dateien benötigen weniger Platz auf Festplatten, SSDs oder Smartphones.
- Dadurch können mehr Fotos, Videos oder Dokumente gespeichert werden.

### 2. Schnellere Datenübertragung
- Kleinere Dateien können schneller über das Internet übertragen werden.
- Webseiten laden schneller und Downloads benötigen weniger Zeit.

### 3. Geringerer Bandbreitenbedarf
- Streaming-Dienste wie Netflix oder YouTube nutzen Komprimierung, um hochauflösende Videos wirtschaftlich übertragen zu können.
- Auch Videokonferenzen und Cloud-Dienste profitieren davon.

### 4. Kosteneinsparungen
- Weniger Speicherplatz und geringerer Datenverkehr reduzieren die Kosten für Unternehmen und Rechenzentren.

## Nachteile und Grenzen

- **Verlustfreie Komprimierung** (z. B. ZIP) spart Speicher, kann die Daten aber nur begrenzt verkleinern.
- **Verlustbehaftete Komprimierung** (z. B. JPEG, MP3) erreicht höhere Kompressionsraten, kann jedoch die Qualität beeinträchtigen.
- Das Komprimieren und Dekomprimieren benötigt Rechenleistung und Energie.

## Bewertung

Ohne Datenkomprimierung wären viele heute selbstverständliche Anwendungen kaum praktikabel:

- Video-Streaming in HD oder 4K
- Musik-Streaming
- Cloud-Speicher
- Messenger mit Bild- und Videoversand
- Videokonferenzen

Daher ist Datenkomprimierung eine Schlüsseltechnologie der digitalen Welt. Der geringe zusätzliche Rechenaufwand wird durch die enormen Einsparungen bei Speicherplatz und Übertragungsvolumen in den meisten Fällen deutlich aufgewogen.
