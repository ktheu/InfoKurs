Das EVA-Prinzip (Eingabe,Verarbeitung,Ausgabe) ist das Grundmodell der Datenverarbeitung.

- Eingabe: Daten, Befehle oder Signale gelangen in das System.
- Verarbeitung: Das System (z.B. der Prozessor/CPU) interpretiert, berechnet und speichert die eingegebenen Daten.
- Ausgabe: Die verarbeiteten Daten werden dem Nutzer oder einem anderen System zugänglich gemacht.

#### Von-Neumann Architektur 

Der Prozessor (CPU, Central Processing Unit) besteht aus Rechenwerk und Steuerwerk. Das Steuerwerk (CU, Control Unit) 
enthält die Elektronik, die es ermöglicht, dass einzelne Computer-Befehle tatsächlich die richtigen Abläufe bewirken. Das Rechenwerk (ALU Arithmetic Logic Unit) enthält die Elektronik, die es möglich macht, dass ein Computer rechnen und logische Verknüpfungen realisieren kann.

(Bilder und Text aus Wikipedia)

Ein Von-Neumann-Rechner beruht auf folgenden Komponenten, die bis heute in Computern verwendet werden:

- **ALU** (Arithmetic Logic Unit) – Rechenwerk, 
 führt Rechenoperationen und boolesche Verknüpfungen aus. 
- **Control Unit** – Steuerwerk oder Leitwerk, 
interpretiert die Anweisungen eines Programms und verschaltet dementsprechend die übrigen Komponenten, es regelt auch die Befehlsabfolge.
- **BUS** – Bus System, 
dient zur Kommunikation zwischen den einzelnen Komponenten (Steuerbus, Adressbus, Datenbus)
- **Memory** – (RAM/Arbeitsspeicher) Speicherwerk, 
speichert sowohl Programme als auch Daten, welche für das Rechenwerk zugänglich sind.
- **I/O Unit** – Eingabe-/Ausgabewerk,
steuert die Ein- und Ausgabe von Daten, zum Anwender (Tastatur, Bildschirm) oder zu anderen Systemen (Schnittstellen).

##### Microbefehle der Johnny-Registermaschine

```text
db->ram    Inhalt des Datenbus im RAM speichern
db->ins    Inhalt des Datenbus ins Instruction Register übernehmen
db->acc    Inhalt des Datenbus im Akkumulator speichern
plus       Akkumulator um Inhalt des Datenbus erhöhen
minus      Akkumulator um Inhalt des Datenbus erniedrigen
ram->db    Inhalt des RAM auf den Datenbus legen

pc->ab     Programm Counter auf den Adressbus legen
pc++       Programm Counter um 1 erhöhen
=0:pc++    Falls Akkumulator = 0, Programm Counter um 1 erhöhen

ins->mc    An die Stelle des Microcodes springen, die im ersten Teil des Instruction Registers steht
ins->pc    Den letzten Teil des Instruction Registers in den Programm Counter schreiben
ins->ab    Den letzten Teil des Instruction Registers auf den Adressbus legen

acc->db    Inhalt des Akkumulators auf den Datenbus legen
acc++      Inhalt des Akkumulators um 1 erhöhen
acc--      Inhalt des Akkumulators um 1 erniedrigen
acc=0      Akkumulator auf 0 setzen

mc=0       Positioniere an die Stelle 0 des Microcodes (FETCH)
stop       Beende das Programm
```

##### Manuelle Microbefehle

Datenbus und Adressbus sind keine Speicher, nur im Modell verhalten sie sich wie eine Art Zwischenspeicher. Die beiden Funktionen ManuellAb und ManuellDb sollen uns das Kennenlernen von Johnny erleichtern. Sie stehen nicht als Befehle für den Microcode zur Verfügung. 


    ManuellAb : eine Zahl auf den Adressbus schreiben
    ManuellDb : eine Zahl auf den Datenbus schreiben

##### Assemblerbefehle der Johnny-Registermaschine

Die Abfolge der Microbefehle für die Assemblerbefehle steht im Microcode.

```text
TAKE x    Kopiert den Wert aus Speicherzelle x in den Akkumulator
ADD x     Akkumulator um den Inhalt des Registers x erhöhen
SUB x     Akkumulator um den Inhalt des Registers x erniedrigen
SAVE x    Inhalt des Akkumulators im Register x speichern

JMP x     Springe zu Register x
TST x     Falls Register x = 0, springe ins übernächste Register

INC x     Erhöhe den Inhalt des Registers x um 1
DEC x     Erniedrige den Inhalt des Registers x um 1
NULL x    Setze den Inhalt des Registers x auf 0

HLT x     Das Programm hält an
```

Uns stehen die Speicherstellen 0 - 999 zur Verfügung.
Mit dem linken Pfeil haben wir die Möglichkeit an bestimmte Speicherstellen im RAM Zahlen von 0 bis 19999 zu speichern. 

Die Zahlen im RAM werden uns im Format XX.YYY angezeigt. Die beiden Ziffern XX entscheiden, ob der Inhalt als Daten oder als Instruktion
interpretiert wird. Falls die Speicherstelle eine Instruktion ist, wird die Bedeutung der Instruktion als Assembler-Code 
angezeigt.

Beispiele:

    00.128 - Daten, hier die Zahl 128
    01.014 - TAKE 14 - Inhalt des Registers 14 in den Accumulator schreiben
    02.010 - ADD 10 - Accumulator um Inhalt des Registers 10 erhöhen
    03.011 - SUB 11 - Accumulator um Inhalt des Registers 11 erniedrigen
    04.011 - SAVE 11 - Inhalt des Accumulators im Register 11 speichern
    05.014 - JMP 14 - Springe zu Register 14
    06.005 - TST 5  - Falls Register 5 = 0, springe in das übernächste Register
    07.010 - INC 10 - Erhöhe Inhalt des Registers 10 um 1
    08.010 - DEC 10 - Erniedrige Inhalt des Registers 10 um 1
    09.012 - NULL 12 - Setze Inhalt des Registers 12 auf 0
    10.011 - HLT 11 - Das Programm hält an. Die Adresse spielt keine Rolle.
