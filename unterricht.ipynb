```python
''' Person
Erstelle eine Klasse Person 
- Attribute: name, alter
- Methoden: vorstellen

Beispiel:

p1 = Person("Anna", 15)
print(p1.vorstellen())

Erwartete Ausgabe:
Hallo, ich heiße Anna und bin 15 Jahre alt.
'''
class Person:
    def __init__(self,name,alter):
        self.alter = alter
        self.name = name

    def vorstellen(self):
        return f'Hallo, ich heiße {self.name} und ich bin {self.alter} Jahre alt.'
```

```python
p1 = Person("Anna", 15)
print(p1.vorstellen())
```
