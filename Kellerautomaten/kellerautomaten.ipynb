### Kellerautomaten

[Video](https://youtu.be/-aVF00KVMns)

<p> Endliche Automaten stoßen an ihre Grenzen, wenn sie sich beliebig viel merken müssten. Ein Beispiel ist die Sprache wie  \( L = \{0^n1^n | n \ge 0\} \) Hier muss der Automat „wissen“, wie viele Nullen er gesehen hat, um später genau gleich viele Einsen zu überprüfen. Das kann ein endlicher Automat nicht – ihm fehlt das Gedächtnis. </p>

Ein Kellerautomat ist wie ein endlicher Automat, aber mit einem extra Speicher: dem Keller (Stack, Stapel). Er kann in den Keller oben etwas ablegen (push) und das oberste Element entfernen (pop). Es ist immer nur das oberste Element sichtbar.

Ein Kellerautomat kann gleichzeitig:

- ein Eingabezeichen lesen
- den Zustand wechseln
- etwas auf den Keller legen oder vom Keller entfernen


<p> Er kann sich durch den Keller mehr merken als ein endlicher Automat. Beispiel:  \( L = \{0^n1^n | n \ge 0\} \). 
Für jede gelesene 0 lege ein Symbol in den Keller, für jede gelesene 1 entferne ein Symbol aus dem Keller.
Am Ende muss der Keller wieder „leer“ sein.
</p>


<h4>Kellerautomat</h4>

<p>Ein <em>nicht-deterministischer Kellerautomat (NPDA = non deterministic pushdown automaton) </em> besteht aus :

<ol>
  <li> \( Q \), einer endlichen Menge von Zuständen</li>
  <li> \( \Sigma \), einem endlichen Eingabe-Alphabet</li>
  <li> \( \Gamma \), einem endlichen Keller-Alphabet</li>
  <li> \( \delta \), einer Übergangsfunktion</li>
  <li> \( s \in Q \), einem Startzustand</li>
  <li> \( \# \in \Gamma \), ein ausgezeichnetes Kellersymbol, das die untere Kellerbegrenzung beschreibt.</li>
  <li> \( F \subseteq Q\), einer Menge von Endzuständen</li>

</ol>
 </p>

Die Übergangsfunktion besteht aus Übergängen der Art: $(q, a, X) \rightarrow (q', \gamma)$. 

Das lesen wir so: „Wenn ich im Zustand $q$ bin und das Zeichen $a$ lese und oben im Keller $X$ liegt, dann gehe ich in den Zustand $q'$, und ersetze $X$ durch $\gamma$.  ($\gamma$ ist ein Zeichenfolge aus dem Keller-Alphabet $\Gamma$)


#### Beispiele



- $(q,0,\#)→(q,0\#)$  - lese 0 und lege 0 auf den Keller
- $(q,1,0)→(q,ε)$ - lese 1 und entferne 0 vom Keller
- $(q,ε,X)→(q′,X)$ - wechsle den Zustand ohne etwas zu lesen

Der Automat akzeptiert ein Wort, wenn es eine Möglichkeit gibt, nach der Abarbeitung des Wortes in einen Endzustand zu gelangen. Bei der Abarbeitung verfolgen wir den Zustand und den Inhalt des Kellers.

Der Automat akzeptiert alle Worte, die mit mindestens einem a beginnen,  gefolgt von b's, wobei die Anzahl der b's größer sein muss als die Anzahl der a's.

Die Zeichen des Kelleralphabets können andere Zeichen als die des Eingabealphabets sein. Manchmal unterscheidet man die beiden Alphabete, indem man für das Eingabealphabet kleine und das Kelleralphabet große Buchstaben verwendet.

#### Infos zu Kellerautomaten

- Die nicht-deterministischen Kellerautomaten (NPDA) erkennen genau die kontextfreien Sprachen.
- Deterministische Kellerautomaten (PDA) erkennen nur einen Teil der kontextfreien Sprachen.

  Die Sprache $L = \{0^n1^n | n \ge 0\}$ kann auch von einem deterministischen Kellerautomaten erkannt werden, aber die Spiegelsprache
  $L = \{ww^R | w \in \{0,1\}^*\}$ nicht. Der Automat müsste "raten", wo die Mitte ist. Deterministische Kellerautomaten können nur Sprachen erkennen, bei denen sie nie raten müssen.
- PDAs mit zwei Stacks sind noch mächtiger. Sie sind äquivalent zu den Turingmaschinen. Mit zwei Kellern kann man das Band einer Turingmaschine simulieren. Keller 1 speichert alles links vom Lesekopf, Keller 2 speichert alles rechts vom Lesekopf. Damit kann man alle berechenbaren (=rekursiv aufzählbaren) Sprachen erkennen.
