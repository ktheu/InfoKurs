### Die Mandelbrotmenge

Jeder Zahl c wird eine Folge von Zahlen zugeordnet.

<img src='mandel10.png' width='601'>

```python
# rekursive Variante
def mandel(c, k):
    '''
    c: Zahl, k: int
    returns: das k-te Element der Mandelbrot-Folge zur Zahl c 
    '''
    if k == 0: return 0
    return mandel(c,k-1)**2 + c  
```

```python
mandel(0.24,5)
```

Für die Berechnung der Mandelbrotfolge nutzen wir die iterative Variante.

```python
def mandelfolge(c, k=10):
    '''
    c: Zahl, k: int
    returns: Liste mit den ersten k Elementen der Mandelbrot-Folge zur Zahl c 
    '''
    a = []
    z = 0
    for i in range(k):
        a.append(z)
        z = z**2 + c
    return a
        
```

```python
mandelfolge(0.24)
```

Es gilt: Wird ein Folgenglied betragsmäßig größer als 2, dann 'explodiert' die Folge (die Folgenelemente gehen betragsmäßig gegen Unendlich).

```python
mandelfolge(0.5)
```

**Eine Zahl c gehört zur Mandelbrotmenge, wenn die Elemente der zu c gehörenden Folge beschränkt bleiben, also betragsmäßig kleiner gleich 2 bleiben.**

```python
print(mandelfolge(-2))
print(mandelfolge(-1))
print(mandelfolge(0))
print(mandelfolge(0.25))
print(mandelfolge(0.5))
```

Bei manchen Zahlen dauert es länger, bis wir feststellen, dass sie nicht zur Mandelbrotmenge gehören.

```python
mandelfolge(0.26,35)
```

Uns interessiert, ab welchem Folgenglied die Elemente betragsmäßig größer als 2 werden (falls das überhaupt passiert)

```python
def mandelzahl(c, kmax):
    '''
    c: Zahl, kmax: int, maximaler Index, bis zu dem geprüft wird
    returns: Index des ersten Folgenglieds, das betragsmäßig größer als 2 ist
             -1 falls alle kmax-Folgenglieder betragsmäßig <= 2
    '''
    a = []
    z = 0
    for i in range(kmax):
        if abs(z) > 2: return i
        a.append(z)
        z = z**2 + c        
    return -1
    
```

Bei 0.25 ist offenbar eine Grenze

```python
c = 0.24
diff = 0.001
for _ in range(20):
    print(f'{c:2.3f} {mandelzahl(c,300)}')
    c+=diff
    
```

Bei -2 ist eine weitere Grenze

```python
c = -1.99
dc = 0.001
for _ in range(20):
    print(f'{c:2.3f}  {mandelzahl(c,300)}')
    c-=dc
```

**Die reellen Zahlen im Intervall $[-2,0.25]$ gehören zur Mandelbrotmenge.**

## Komplexe Zahlen

Eine komplexe Zahl stellen wir uns vor als Punkt in der Ebene. Statt der Koordinaten $(a|b)$ schreiben wir $a + bi$.
Die Zahl $a$ nennen wir den Realteil, $b$ den Imaginärteil der komplexen Zahl.

<img src="mandel1.png" width="500">

Der Betrag einer komplexen Zahl ist ihr Abstand zum Nullpunkt.

<img src="mandel1a.png" width="500">

```python
# In Python wird der Imaginärteil durch ein nachgestelltes j gekennzeichnet
z = 3 + 2j
abs(z)
```

Zwei komplexe Zahlen werden addiert, indem die Realteile und die Imaginärteile addiert werden. Dies entspricht der
normalen Vektoraddition.

<img src="mandel2.png" width="501">

```python
z1 = 3 + 2j
z2 = -2 + 1j
z1 + z2
```

Zwei komplexe Zahlen werden multipliziert, indem die Beträge multipliziert und die Winkel addiert werden. Inbesondere gilt dann: $i^2 = -1$

<img src="mandel3.png" width="651">

```python
1j ** 2
```

```python
z1 = 3 + 2j
z2 = -2 + 1j
z1*z2
```

Für das Quadrat einer komplexen Zahl gilt:
$(a + bi)^2 = a^2-b^2 + 2abi$

#### Übung:

$z_1 = 3 - i, z_2 = 3 + 2i$. Berechne

- $z_1 + z_2$
- $z_1 * z_2$
- $z_1^2 + z_2$
 

```python
z1 = 3 - 1j
z2 = 3 + 2j
z1+z2, z1*z2, z1**2+z2
```

Die numpy-Funktion linspace teilt ein Intervall in gleichmäßige Abschnitte ein

Mit enumerate können wir uns den Index und den Wert einer Aufzählung geben lassen.

```python
N = 10
Nx = 600                           # Anzahl Punkte in x-Richtung
Ny = int(Nx * hoehe/breite)        # Anzahl Punkte in y-Richtung

img = Image.new('RGB', (Nx,Ny), "white") 
pixels = img.load()  

for xi,x in enumerate(np.linspace(x0,x1,Nx)):
    for yi,y in enumerate(np.linspace(y0,y1,Ny)):
        z = x + 1j*y
        if mandelzahl(z,N) < 0:
            pixels[xi,yi] = 0   

display(img)  
```

Wir färben die nicht-schwarzen Punkte mit Farben ein, abhängig davon, bei welcher Iteration der Punkt als nicht zur Mandelbrotmenge
gehörig entdeckt wurde.


```python
farben = [(232, 232, 219),(255, 223, 101),(133, 131, 119),(99, 96, 87),(73, 72, 70)];
N = 30
Nx = 600                           # Anzahl Punkte in x-Richtung
Ny = int(Nx * hoehe/breite)        # Anzahl Punkte in y-Richtung

img = Image.new('RGB', (Nx,Ny), "white") 
pixels = img.load()  
 

import numpy as np
for xi,x in enumerate(np.linspace(x0,x1,Nx)):
    for yi,y in enumerate(np.linspace(y0,y1,Ny)):
        if xi == 100 and yi == 50:
            print(x,y)
        z = x + 1j*y
        mz = mandelzahl(z,N)
        if mz < 0:
            pixels[xi,yi] = 0
        else:
            pixels[xi,yi] = farben[mz%len(farben)]

display(img)  
```

In den Numberphile-Videos wird die Mandelbrot-Folge wie folgt definiert:

Für jede Zahl c definieren wir eine Funktion $f_c(z) = z^2 + c$ und betrachten das Verhalten von 0 bei Iteration ("the behaviour of 0 under iteration"). D.h. wir beobachten, was passiert, wenn 
man die 0 in diese Funktion einsetzt und dann das Resultat wieder in die Funktion einsetzt, dann das Resultat wieder in die Funktion einsetzt usw. Dies führt zur Folge $$c , c^2 + c, (c^2 + c)^2 + c, ...$$  

"A number c is in the mandelbrot set if the function $f_c$ does not blow up under iteration of 0"
