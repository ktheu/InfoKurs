### Tiefensuche

Die Tiefensuche besucht jeden Knoten genau einmal. Wir besuchen die Knoten rekursiv:

    def dfs(v): 
        merke v als besucht
        für alle Nachbarn w von v:
            wenn w nicht besucht:
                dfs(w)

**Vereinbarung** (für einheitliche Ergebnisse): wir besuchen Knoten in alphabetischer Reihenfolge.

```python
def dfs(v):  
    print(v)
    visited[v] = True
    for w in G[v]:
        if not visited[w]:
            dfs(w) 

visited =  {v : False for v in G}  
for v in G:
    if not visited[v]:
        dfs(v)
 
```

### Topologische Sortierung

Wir wollen eine Nummerierung der Knoten finden, die mit den Kantenrichtungen
kompatibel ist.

Genau die **gerichteten azyklischen Graphen (directed acyclic graph, DAG**) können topologisch sortiert werden. 

##### Frage:

Welchen Zusammenhang gibt es zwischen den pre-/post-visit Nummern und einer topologischen Sortierung. 

Die abwärts sortierte Reihe der postvisit-Nummern ergibt eine topologische Sortierung.

    ['f', 'a', 'c', 'e', 'h', 'b', 'd', 'i', 'g', 'j']

#### topoSort

```python
def topoSort(G):
    def dfs(v):
        nonlocal counter
        visited[v] = True
        for w in G[v]:
            if not visited[w]:
                dfs(w)
        postvisit[v] = counter
        counter += 1
    
    visited = {v : False for v in G}
    postvisit = {v : 0 for v in G}
    counter = 1
    
    for v in G:
        if not visited[v] :
            dfs(v)

    return sorted(G.keys(),key=lambda v: postvisit[v],reverse = True)
        
```

### Backtracking 

Backtracking ist ein systematisches Ausprobieren. Wenn wir in eine Sackgasse geraten sind, gehen wir zurück (backtrack) und probieren die nächste Variante aus. Meistens sind alle möglichen Lösungen gefragt.

Backtracking ist anwendbar, wenn wir eine Lösung schrittweise aufbauen können und wenn wir für eine
Teillösung die möglichen Kandidaten aufzählen können, die die Teillösung erweitern. 

Eine Teillösung bezeichnen wir mit $a$. $a$ ist eine Liste mit Entscheidungen $a[0]...a[k]$, die zur Lösung führen sollen.

Beim Backtracking wird eine Teillösung immer wieder verändert. Erst wenn eine Lösung gefunden wird, wird eine Kopie gesichert.

##### Schablone für das Backtracking:

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            back(a)
            a.pop()
                
def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    pass

def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern
    '''
    pass

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    pass

solutions = []
back([])
print(solutions)
```

Wir fassen den Code der Schablone zusammen:

```python
# Permutationen
def permutation(n):
    def back(a):
        if len(a) == n:
            solutions.append(a.copy())
        for k in range(1,n+1):
            if k not in a:
                a.append(k)
                back(a)
                a.pop()

    solutions = []
    back([])
    return solutions
```

#### Beispiel 2

Wir suchen alle (einfachen) Kreise in einem Graphen G, die an einem Knoten v starten. Einfache Kreise sind Pfade, bei denen nur Anfangs- und Endpunkt gleich sind (die keine inneren Kreise haben).

```python
def back(a):
    if isSolution(a):
        solutions.append(a.copy())
    for cand in candidates(a):
        if isGood(cand,a):
            a.append(cand)
            back(a)
            a.pop()
                
def isSolution(a):
    '''
    returns: True, wenn die Teillösung a eine Lösung darstellt
    '''
    return len(a) > 1 and a[-1] == a[0]

def candidates(a):
    '''
    returns: Liste von Entscheidungen, die die Teillösung a um eine Stufe erweitern
    '''
    return G[a[-1]]

def isGood(cand, a):
    '''
    returns: True, wenn die Entscheidung cand die Teillösung a sinnvoll erweitert.
    '''
    return cand not in a[1:] 
```

```python
G = {'a':{'b'}, 'b':{'c'}, 'c':{'d','f'}, 'd':{'e'}, 'e':{'f'}, 'f':{'a','c','e'}}
solutions = []
back(['f'])
print(solutions)
```

Wir fassen den Code wieder zu einer Funktion zusammen:

```python
def kreise(G, v):
    def back(a):
        if len(a) > 1 and a[-1] == a[0]:
            solutions.append(a.copy())
        for cand in G[a[-1]]:
            if cand not in a[1:]:
                a.append(cand)
                back(a)
                a.pop()

    solutions = []
    back([v])
    return solutions
    
```

Kreise sind eindeutig durch die Menge ihrer Kanten. 

```python
def getKanten(kreis):
    tmp = set()
    for i in range(len(kreis)-1):
        tmp.add(kreis[i]+kreis[i+1])
    return tmp
 
kreis =  ['f', 'a', 'b', 'c', 'd', 'e', 'f']
getKanten(kreis)
```

Jetzt können wir die verschiedenen Kreise in dem Graphen ermitteln.

```python
def alleKreise(G):
    tmp = []
    for v in G:
        for kreis in kreise(G,v):
            kanten = getKanten(kreis)
            if kanten not in tmp:
                tmp.append(kanten)
    return tmp
```
