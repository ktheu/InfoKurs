```python
def makeGraph(gs, directed=True):
    '''
    >>> gs = 'ab ac cb cd db bd'
    >>> makeGraph(gs)
    {'a': {'b', 'c'}, 'b': {'d'}, 'c': {'b', 'd'}, 'd': {'b'}}
    '''
    a = gs.split()
    V = {c for c in gs if c != ' '}
    G = {v: set() for v in sorted(V)}
    for s in a:
        v, w = s[0], s[1]
        G[v].add(w)
        if not directed:
            G[w].add(v)
    return G


def printEdges(G):
    '''
    G: Graph in der Form
    {'a': {'b', 'c'}, 'b': {'d'}, 'c': {'b', 'd'}, 'd': {'b'}}
    returns 
    '''
    for u in G:
        for v in G[u]:
            print(u,v)


```

```python
def getNodesEdgesFrom(G, weighted=False):
    '''
    G: Graph
    returns: list of nodes, list of edges
    '''
    nodes = list(G.keys())
    edges = []
    if weighted:
        for u in G:
            for v in G[u]:
                edges.append((u,v,G[u][v]))
    else:
        edges = [(u,v) for u in G for v in G[u]]
    return nodes, edges
```

```python
G = {'a':{'b','c'}, 'c':{'b','d'}, 'b':{'d'}, 'd':{'b'}}
nodes, edges = getNodesEdgesFrom(G)
```

```python
import networkx as nx
import numpy as np
from matplotlib import pyplot as plt

G = nx.DiGraph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)

plt.figure(figsize=(3, 2))
np.random.seed(1)

nx.draw(G, with_labels=True, arrows=True, node_color='skyblue',arrowsize=20)
```
