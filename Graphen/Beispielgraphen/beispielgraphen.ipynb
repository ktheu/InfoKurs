## Graphen - Beispiele

Graphen erzeugt mit: https://csacademy.com/app/graph_editor/ 

```python
def graphEditorInput(G):
    '''
    G: Graph in der Form
    {'a': {'b', 'c'}, 'b': {'d'}, 'c': {'b', 'd'}, 'd': {'b'}}
    returns: Input für https://csacademy.com/app/graph_editor/ 

    '''
    def getKnoten(G):
        V = set()
        for u in G:
            V.add(u)
            for v in G[u]:
                V.add(v)
        return sorted(V)

    def getKanten(G):
        edges = []
        for u in sorted(G):
            for v in G[u]:
                edges.append((u,v))
        return sorted(edges)
    
    V = getKnoten(G)
    edges = getKanten(G)
    for u in V:
        print(u)
    for u, v in edges:
        print(u,v)

def weightedGraphEditorInput(G):
    '''
    G: Graph in der Form
    {'a': {'b': 3, 'c': 2}, 'b': {}, 'c': {'b': 2, 'd': 1}, 'd': {}}
    returns: Input für https://csacademy.com/app/graph_editor/ 
    '''
    def getKnoten(G):
        V = set()
        for u in G:
            V.add(u)
            for v in G[u]:
                V.add(v)
        return sorted(V)

    def getKanten(G):
        edges = []
        for u in G:
            for v in G[u]:
                edges.append((u,v,G[u][v]))
        return sorted(edges)
    
    V = getKnoten(G)
    edges = getKanten(G)
    for u in V:
        print(u)
    for u, v, cost in edges:
        print(u,v,cost)

def makeGraph(gs, directed=True):
    '''
    >>> gs = 'ab ac cb cd db bd'
    >>> makeGraph(gs)
    {'a': {'b', 'c'}, 'b': {'d'}, 'c': {'b', 'd'}, 'd': {'b'}}
    '''
    a = gs.split()
    V = {c for c in gs if c != ' '}
    G = {v: set() for v in sorted(V)}
    for s in a:
        v, w = s[0], s[1]
        G[v].add(w)
        if not directed:
            G[w].add(v)
    return G
    
def makeWeightedGraph(gs, directed=True):
    '''
    >>> gs = 'ab8 ac2 cb1 cd6 db2 bd4'
    >>> makeGraph(gs)
    {'c': {'b': 1, 'd': 6}, 'b': {'d': 4}, 'a': {'b': 8, 'c': 2}, 'd': {'b': 2}}
    '''
    V = set([c for c in gs if (not c.isdigit() and c != ' ')])
    G = {v: dict() for v in V}
    for s in gs.split():
        u,v,val = s[0],s[1],int(s[2:])
        G[u][v] = val

        if not directed:
            G[v][u] = val
    return G
 
```
