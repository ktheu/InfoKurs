Ein **Graph** besteht aus **Knoten** (vertices) und **Kanten** (edges). Er kann gerichtet oder ungerichtet sein. Die Kanten 
können **gewichtet** sein (Kosten).  

<img src='graphen_01.png'>
<img src='graphen_02.png'>


Mit Graphen können binäre Beziehungen zwischen Objekten modelliert werden. Die Objekte werden
durch die Knoten, die Beziehungen durch die Kanten modelliert.  

* Orte: Entfernungen, Dauer
* Personen: ist befreundet, verheiratet mit
* Ereignisse: a muss vor b geschehen

#### Begriffe

<img src='graphen40.png' width='600'>
 

* x ist zu y und z adjazent, x hat die Nachbarn y und z
* a hat die Nachbarn b, c, d
* der Grad von x ist 2
* der Eingangsgrad von a ist 1, der Ausgangsgrad ist 3
* a ist Vorgänger von c, c ist Nachfolger von a
* Ein Weg ist eine Folge von adjazenten Knoten, z.B: y-x-z oder a-c-d-b
* Ein Kreis ist ein Weg, der zurück zum Startknoten führt, z.B. a-c-d-b-a

#### Der Algorithmus von Floyd-Warshall

Der Algorithmus von Floyd-Warshall löst das **all-pairs-shortest-path**-Problem

<img src='graphen41.png' width='1001'>

<img src='graphen_09.png'>

```python
def printMatrix(a):
    for i in range(len(a)):
        for j in range(len(a)):
            if a[i][j] == inf:
                print("{:4}".format("."),end='')
            else:
                print("{:<4}".format(a[i][j]),end='')
        print()
    print()

G = [[0,   8,   2, inf],
     [inf, 0, inf,   4],
     [inf, 1,   0,   6],
     [inf, 2, inf,   0]]

printMatrix(G)
```

```python
n = len(G)
D = [[0]*n for j in range(n)]   # Matrix D
for i in range(n):
    for j in range(n):
        D[i][j] = G[i][j]

for k in range(n):
    for i in range(n):
        for j in range(n):
            if D[i][k] + D[k][j] < D[i][j]:
                D[i][j] = D[i][k] + D[k][j]

printMatrix(D)
```

Um den kürzesten Weg zu rekonstruieren, benötigen wir eine Matrix $P$, die an Position 
$P[i][j]$ den vorletzten Knoten auf dem kürzesten Weg von i nach j notiert.

<img src='graphen43.png' width='700'>

Dazu bilden wir eine Folge von Matrizen $P^k$, die an Position 
$P^k[i][j]$ den vorletzten Knoten auf dem kürzesten Weg von i nach j notiert, der als Zwischenknoten höchstens 
$0,1,..., k$ nutzt.

```python
n = len(G)
D = [[0]*n for j in range(n)]   # Matrix D
P = [[0]*n for j in range(n)]   # Matrix P
for i in range(n):
    for j in range(n):
        D[i][j] = G[i][j]
        P[i][j] = i

for k in range(n):
    for i in range(n):
        for j in range(n):
            if D[i][k] + D[k][j] < D[i][j]:
                D[i][j] = D[i][k] + D[k][j]
                P[i][j] = P[k][j]

printMatrix(D)
printMatrix(P)
```

```python
def getPath(P, i, j):
    if i == j:  return str(i)
    return getPath(P,i,P[i][j]) + '-' + str(j)

getPath(P,0,3)
```

```python
def floyd(G):
    n = len(G)
    D = [[0]*n for j in range(n)]   # Matrix D
    P = [[0]*n for j in range(n)]   # Matrix P
    for i in range(n):
        for j in range(n):
            D[i][j] = G[i][j]
            P[i][j] = i
    
    for k in range(n):
        for i in range(n):
            for j in range(n):
                if D[i][k] + D[k][j] < D[i][j]:
                    D[i][j] = D[i][k] + D[k][j]
                    P[i][j] = P[k][j]
    return D, P
 
inf = float('inf')
G = [[0,   8,   2, inf],
     [inf, 0, inf,   4],
     [inf, 1,   0,   6],
     [inf, 2, inf,   0]]

D, P = floyd(G)

print("Gegebenen Kostenmatrix:")
printMatrix(G)

print("Errechnete Distanzmatrix:")
printMatrix(D)

print("Errechnete Wegematrix:")
printMatrix(P)

start = 2
ziel = 3
print(f"Kürzester Weg von {start} nach {ziel}: {getPath(P,start,ziel)} mit Kosten {D[start][ziel]}. ")
```

Laufzeit des Floyd-Warshall-Algorithmus: $O(|V|^3)$

### Erreichbarkeit

Gegeben ein Graph G und ein Startknoten s. Welche Knoten sind von s erreichbar?

<img src='graph6z.png' width='301'>




Die Tiefensuche (depth first search, dfs) besucht jeden Knoten genau einmal. Wir besuchen die Knoten rekursiv:

    def dfs(v): 
        merke v als besucht
        für alle Nachbarn w von v:
            wenn w nicht besucht:
                dfs(w)

Vereinbarung (für einheitliche Ergebnisse bei manueller Durchführung der Algorithmen): Wir besuchen Knoten in alphabetischer Reihenfolge.

```python
G = {'a': set('bc'),
     'b': set('d'),
     'c': set('ed'),
     'd': set('g'),
     'e': set(),
     'f': set('d'),
     'g': set('e')}  
```

```python
def dfs(v):  
    visited[v] = True
    for w in G[v]:
        if not visited[w]:
            dfs(w) 

visited =  {v : False for v in G}  
dfs('a')

for v in G:
    if visited[v]:
        print(v, end=' ')
```

Laufzeit: Jeder Knoten wird höchstens einmal besucht $O(|V|)$, jeder Knoten prüft seine Nachbarn. Die Zahl der Nachbarn liegt in $O(|E|)$, also insgesamt $O(|V|+|E|)$

### Zusammenhangskomponenten eines ungerichteten Graphen

Die Knoten eines ungerichteten Graphen G können in **Zusammenhangskomponenten** (Connected Components) aufgeteilt werden, so dass v von w genau dann erreichbar ist, wenn beide in derselben Zusammenhangskomponente liegen.

<img src='graph7.png' width='302'>

```python
G = {
    'a': set('e'),
    'b': set('e'),
    'c': set('d'),
    'd': set('c'),
    'e': set('abf'),
    'f': set('e')
}

def dfs(v):  
    visited[v] = True
    cc[v] = ccnum
    for w in G[v]:
        if not visited[w]:
            dfs(w) 

visited =  {v : False for v in G}  
cc = dict()
ccnum = 1 
for v in G:
    if not visited[v]:
        dfs(v)
        ccnum+=1
 
for i in range(1,ccnum):
    result = [v for v in G if visited[v] and cc[v] == i]
    print(i,'-',*result)
```

### Topologische Sortierung

Wir wollen eine Nummerierung der Knoten finden, die mit den Kantenrichtungen
kompatibel ist.

<img src='graph10.png' width='500'>

Es gilt: ein Graph kann genau dann topologisch sortiert werden, wenn er keinen Kreis enthält.
Genau die **gerichteten azyklischen Graphen (directed acyclic graph, DAG**) können topologisch sortiert werden. 

Eine **Quelle (source)** ist ein Knoten mit dem Eingangsgrad 0. Eine **Senke (sink)** ist ein Knoten mit dem Ausgangsgrad 0.


Idee für die topologische Sortierung (1. Versuch):

    Solange der Graph nicht leer:
        Folge einem Pfad bis es nicht mehr weitergeht.
        Finde dort eine Senke
        Füge die Senke in die Sortierung wie in einen Stapel ein
        Entferne die Senke aus dem Graphen.

Laufzeit:  $O(\left|V\right|^2)$    -   nicht gut

Verbesserungsmöglichkeit: statt den Pfad immer wieder von vorne zu beginnen, wird nach der
Entfernung der Senke der Weg so weit wie nötig zurückgegangen. Das ist Tiefensuche! Die abwärts sortierte Reihe der postvisit-Nummern ergibt eine topologische Sortierung.

<img src='graph10n.png' width='500'>

```python
G = {
'a': set('c'),
'b': set('c'),
'c': set('d'),
'd': set('g'),
'e': set('df'),
'f': set('g'),
'g': set()
}

def dfs(v):
    global counter
    visited[v] = True
    for w in G[v]:
        if not visited[w]:
            dfs(w)
    postvisit[v] = counter
    counter += 1
    
visited = {v : False for v in G}
postvisit = dict()
counter = 1

for v in G:
    if not visited[v] :
        dfs(v)

topolist = sorted(G.keys(),key=lambda v: postvisit[v],reverse = True)
print(topolist)
```

Die **Breitensuche (breadth first search, bfs)** traversiert den Graphen in Schichten nach der Entfernung zum Ausgangsknoten.
Zu jedem Knoten merkt man sich seinen Vorgänger (prev), dadurch entsteht ein **shortest-path Baum**.
(Ein Baum ist ein zusammenhängender Graph, der keine Kreise enthält). Daraus lässt sich der kürzeste
Weg vom Ausgangsknoten zu jedem anderen Knoten rekonstruieren.

Der shortest-path Baum für die Breitensuche ab Knoten 'b':

<img src='graph11_sp_baum.png' width='350'>

    # Breitensuche mit Ausgangsknoten s:
    Für jeden Knoten u in G: 
        dist(u) = unendlich 
        prev(u) = None
    
    Füge s in eine Schlange Q ein 
    dist(s) = 0
    
    Solange Q nicht leer:
        hole Knoten u aus der Schlange
            für alle Nachbarn v von u:
                falls dist(v) unendlich:
                    Füge v in die Schlange Q ein
                    dist(v) = dist(u) + 1
                    prev(v) = u

```python
from collections import deque

G = {
'a': set('bd'),
'b': set('acde'),
'c': set('bf'),
'd': set('abg'),
'e': set('bf'),
'f': set('cei'),
'g': set('dh'),
'h': set('fgi'),
'i': set('fh')
}

def reconstructPath(ziel,prev):
    path = []
    v = ziel
    while v is not None:
        path.append(v)
        v = prev[v]
    path.reverse()
    return path

inf = float('inf')
dist = {v:inf for v in G}
prev = {v:None for v in G}

s = 'b'
dist[s] = 0
Q = deque([s])          
while Q:
    u = Q.popleft()
    for v in sorted(G[u]):       # Sortierung darf fehlen (nur zur Überprüfung)
        if dist[v] == inf:
            Q.append(v)
            dist[v] = dist[u]+1
            prev[v] = u
            
ziel = 'i'
print(f'Entfernung von {s} nach {ziel} ist {dist[ziel]} mit Pfad {reconstructPath(ziel,prev)}')
```
