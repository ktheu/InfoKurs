```python
def makeGraph(gs):
    '''
    >>> gs = 'ab ac cb cd db bd'
    >>> makeGraph(gs)
    {'a': {'b', 'c'}, 'b': {'d'}, 'c': {'b', 'd'}, 'd': {'b'}}
    '''
    a = gs.split()
    V = {c for c in gs if c != ' '}
    G = {v: [] for v in sorted(V)}
    for s in a:
        v, w = s[0], s[1]
        G[v].append(w)

    for v in G: 
        G[v].sort()
    return G

def printEdges(G):
    '''
    G: Graph in der Form
    {'a': {'b', 'c'}, 'b': {'d'}, 'c': {'b', 'd'}, 'd': {'b'}}
    returns 
    '''
    for u in G:
        for v in G[u]:
            print(u,v)

def topoSort(G):
    def dfs(v):
        nonlocal counter
        visited[v] = True
        for w in G[v]:
            if not visited[w]:
                dfs(w)
        postvisit[v] = counter
        counter += 1
    
    visited = {v : False for v in G}
    postvisit = {v : 0 for v in G}
    counter = 1
    
    for v in G:
        if not visited[v] :
            dfs(v)

    return sorted(G.keys(),key=lambda v: postvisit[v],reverse = True)


def prePostVisit(G):
    def dfs(v):
        nonlocal counter
        visited[v] = True
        previsit[v] = counter
        counter += 1
        for w in sorted(G[v]):  #  alphabetische Reihenfolge
            if not visited[w]:
                dfs(w)
        postvisit[v] = counter
        counter += 1
    
    visited = {v : False for v in G}
    previsit = {v : 0 for v in G}
    postvisit = {v : 0 for v in G}
    counter = 1

    for v in G.keys():
        if not visited[v] :
            dfs(v)

    tmp = []
    for v in sorted(G.keys(),key=lambda v: previsit[v]):
        tmp.append((v,previsit[v],postvisit[v]))
    return tmp

def reihenfolge(G):
    tmp = []
    def dfs(v):  
        tmp.append(v)
        visited[v] = True
        for w in G[v]:
            if not visited[w]:
                dfs(w) 
    
    visited =  {v : False for v in G}  
    for v in G:
        if not visited[v]:
            dfs(v)
    return tmp

def kreise(G, v):
    def back(a):
        if len(a) > 1 and a[-1] == a[0]:
            solutions.append(a.copy())
        for cand in G[a[-1]]:
            if cand not in a[1:]:
                a.append(cand)
                back(a)
                a.pop()

    solutions = []
    back([v])
    return solutions

def getKanten(kreis):
    tmp = []
    for i in range(len(kreis)-1):
        tmp.append(kreis[i]+kreis[i+1])
    tmp.sort()
    return tmp

def alleKreise(G):
    tmp = []
    for v in G:
        for kreis in kreise(G,v):
            kanten = getKanten(kreis)
            if kanten not in tmp:
                tmp.append(kanten)
    return tmp
    
```

```python
import random
def randomGraph(anzKnoten, anzKanten):
    abc = 'abcdefghijklmnopqrstuvwxyz'
    V = abc[:anzKnoten]
    edges = set()
    while len(edges) < anzKanten:
        v = random.choice(V)
        w = random.choice(V)
        if v != w:
            edges.add(v+w)
    
    return  ' '.join(list(edges))
 


```
