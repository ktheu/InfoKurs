### Die Komponenten eines relationalen Datenbanksystems

Ein relationales Datenbanksystem besteht aus einer in Tabellen
strukturierten Sammlung von Daten (Datenbanken) und einer 
Software zur Verwaltung dieser Daten, dem sog. Datenbankmanagementsystem (DBMS).
Das DBMS kontrolliert Zugriff und Verarbeitung der Daten in der Datenbasis.
Anwender oder Anwendungsprogramme nutzen die Funktionalität des DBMS über eine Datenbankschnittstelle.

Wie verwenden folgende Begriffe:


- **Entität:** Ein konkretes Objekt der realen Welt, z. B. das Buch „Tschick“.

- **Entitätstyp:** Die Klasse ähnlicher Objekte, hier „Buch“.

- **Attribut:** Eine Eigenschaft eines Entitätstyps, z. B. Titel oder Autor.

- **Tabelle:** Die strukturierte Darstellung eines Entitätstyps in Zeilen und Spalten.

- **Datensatz:** Eine Zeile der Tabelle, die alle Informationen zu einer Entität enthält.

- **Datenfeld:** Ein einzelner Wert innerhalb eines Datensatzes, z. B. „Tschick“ oder „2010“.

#### Relation und Tabelle

Die Grundlagen für die relationalen Datenbanken schuf 1970 der Mathematiker Edgar F. Codd.

Eine Relation besteht aus Attributen und einer Menge von Tupeln. Die obige Tabelle kann mathematisch durch die Relation

$\text{Buch(BuchID,Titel,Autor,Erscheinungsjahr)}$

mit der Tupelmenge

$\{(101,\text{'Die Physiker'},\text{'Dürrenmatt'},1962),(102,\text{'Tschick'},\text{'Herrndorf'},2010),(103,\text{'Krabat'},\text{'Preußler'},1971)\}$

beschrieben werden.

Die Tabelle ist eine übersichtliche Darstellung dieser Relation. Dabei spielt für die Relation weder die Reihenfolge der Zeilen noch die Reihenfolge der Spalten eine Rolle. Die Bedeutung der einzelnen Werte ergibt sich aus den Attributnamen.

Wenn sich keine Spalte oder Spaltenkombination als Schlüssel anbietet führt man einen **künstlichen Schlüssel** ein, meist mit dem Namenszusatz **ID**.

### Entwurf einer Datenbank
1. Wir modellieren unser Anwendungsproblem mit Entitäten und ihren Beziehungen (ER-Modell)
2. Wir setzen das ER-Modell in Tabellen um.
3. Wir reorganisieren die Tabellen um minimale Redundanz und maximale Konsistenz zu erreichen.

### Entity-Relationship Modell

**Entitäten** entsprechen Objekten der realen Welt, **Entitätstypen** werden im
**ER-Diagramm** durch Rechtecke dargestellt. **Attribute** der Entitätstypen werden durch Ellipsen dargestellt.

Eine **Beziehung (Relationship)** beschreibt den Zusammenhang von Entitäten untereinander. Gleichartige Beziehungen fasst man zu **Beziehungstypen** zusammen und stellt sie durch eine Raute dar. Auch Beziehungen können Attribute haben.

**Kardinalitäten** beschreiben, wie viele Entitäten an einer
konkreten Beziehung beteiligt sein können. Es sind 1:1, 1:n und
m:n-Beziehungen möglich.

Für jeden Entitätstyp wird ein **Primärschlüssel** festgelegt, der eine konkrete Entität eindeutig identifiziert. Der Primärschlüssel wird im ER-Diagramm unterstrichen. Der **Primärschlüssel** kann aus einem oder mehreren Attributen bestehen, oft ist er eine künstlich eingeführte **ID**.

Beispiel: ER-Diagramm für einen Online-Buchhandel

<img src='datenbanken01.png'>
 

### Umsetzung des ER-Modells in Tabellen

Die Informationen des ER-Diagramms werden in Tabellen überführt. Die Struktur der Tabellen wird als **Schema** bezeichnet.
Im Schema werden Primärschlüssel unterstrichen und Fremdschlüssel mit einem Pfeil nach oben versehen. 

----

**Jede Entität bekommt eine eigene Tabelle.** 

* Verlag(<u>VID</u>,Name,Sitz,Ansprechpartner)  
* Buch(<u>ISBN</u>,Titel,Autor,Preis,Kategorie)
* Kunde(<u>Benutzername</u>,Passwort,Vorname,Nachname,Adresse,E-Mail)
* Darstellung(<u>DID</u>,d_Bewertung,lieferbar)

**Eine 1:1-Beziehung erhält keine eigene Tabelle.** Einer der beiden an der Beziehung beteiligten Relationen erhält den Primärschlüssel der anderen als Fremdschlüssel. Attribute der Beziehung werden ggf. übernommen.

* Darstellung(<u>DID</u>,d_Bewertung,lieferbar,ISBN↑)

**Eine 1:n Beziehungen erhält keine eigene Tabelle.** Die Relation mit Kardinalität n erhält den anderen Primärschlüssel als Fremdschlüssel. Attribute der Beziehung werden ggf. übernommen.

* Buch(<u>ISBN</u>,Titel,Autor,Preis,Kategorie,VID↑)



**Eine n:m-Beziehung erhält eine eigene Tabelle.** Die Primärschlüssel der beteiligten Entitäten werden als Fremdschlüssel hinzugefügt. Die Kombination der Fremdschlüssel ist der neue Primärschlüssel. Attribute der Beziehung werden ggf. übernommen.

* bestellt(<u>ISBN↑</u>,<u>Benutzername</u>↑,Anzahl)

### Normalisierung

Durch Umstrukturierung kann man eine Datenbank in verschiedene Normalformen bringen: **1NF, 2NF, 3NF, Boyce-Codd NF, 4NF**.
Eine Datenbank in 3NF ist in der Regel frei von Anomalien.

Wir beschränken uns hier auf eine informelle Beschreibung:

**1NF**: Die Attribute sind atomar und es kommen keine Wiederholfelder vor. <br>
**2NF**: 1NF + jedes nicht zum Primärschlüssel gehörende Attribut ist vom gesamten Primärschlüssel abhängig, nicht nur von einem Teil des Schlüssel. <br>
**3NF**: 2NF + es gibt keine Abhängigkeiten zwischen Nichtschlüsselattributen.

### SQLITE

Wir betrachten im Folgenden das DBMS **SQLITE**. Python bietet uns mit dem Modul **sqlite3** eine Schnittstelle, mit dem wir unsere Anweisungen an das DBMS übermitteln können. **SQL** ist die Sprache, in der diese Anweisungen formuliert werden.

#### Connection und Cursor: Verbindung zur Datenbank aufbauen

Die Verbindung (Connection) zur Datenbank gibt uns einen Cursor zurück, über den wir unsere SQL Anweisungen übermitteln.

```python
import sqlite3
con = sqlite3.connect("lkwahl.db")    # wird neu angelegt, falls noch nicht exisitert
cur = con.cursor()

# hier kommen unsere SQL-Anweisungen hin

con.commit()        # falls Änderungen in der Datenbank gespeichert werden sollen
con.close()
```

#### DROP und CREATE: Tabelle löschen und erzeugen

```python
import sqlite3
con = sqlite3.connect("lkwahl.db")    # wird neu angelegt, falls noch nicht exisitert
cur = con.cursor()

cur.execute("DROP TABLE IF EXISTS Lehrer")
cur.execute("""
CREATE TABLE Lehrer (
    Kuerzel TEXT PRIMARY KEY,
    Lehrer TEXT NOT NULL
)
""")

con.commit()
con.close()
```

Die wichtigsten Datentypen beim Anlegen einer Tabelle sind für uns:

    INTEGER, REAL, TEXT

Das entspricht den Python-Datentypen

    int, float, str

Wir legen die 4 Tabellen aus dem Beispiel für die LK-Wahl an. Wenn wir in einem execute mehrere SQL-Befehle absetzen, müssen wir diese mit einem Strichpunkt trennen.

```python
import sqlite3
con = sqlite3.connect("lkwahl.db")    # wird neu angelegt, falls noch nicht exisitert
cur = con.cursor()

cur.execute("""

CREATE TABLE Schueler (
    SNr INTEGER PRIMARY KEY,
    Name TEXT,
    Vorname TEXT
);

CREATE TABLE Lehrer (
    Kuerzel TEXT PRIMARY KEY,
    Lehrer TEXT
);

CREATE TABLE Kurse (
    KID INTEGER PRIMARY KEY,
    Kuerzel TEXT,
    FOREIGN KEY (Kuerzel) REFERENCES Lehrer(Kuerzel)
);

CREATE TABLE LKWahl (
    SNr INTEGER,
    KID INTEGER,
    PRIMARY KEY (SNr, KID),
    FOREIGN KEY (SNr) REFERENCES Schueler(SNr),
    FOREIGN KEY (KID) REFERENCES Kurse(KID)
);

""")

con.commit()
con.close()
```

### SQL-Beispiele

Zunächst bauen wir die Datenbank auf

```python
import sqlite3

# Datenbank anlegen bzw. öffnen
con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

# Vorhandene Tabellen löschen
cur.execute("DROP TABLE IF EXISTS LKWahl")
cur.execute("DROP TABLE IF EXISTS Schueler")
cur.execute("DROP TABLE IF EXISTS Kurse")
cur.execute("DROP TABLE IF EXISTS Lehrer")

# Tabelle Lehrer
cur.execute("""
CREATE TABLE Lehrer (
    Kuerzel TEXT PRIMARY KEY,
    Lehrer TEXT NOT NULL
)
""")

# Tabelle Kurse
cur.execute("""
CREATE TABLE Kurse (
    KID TEXT PRIMARY KEY,
    Kuerzel TEXT NOT NULL,
    FOREIGN KEY (Kuerzel) REFERENCES Lehrer(Kuerzel)
)
""")

# Tabelle Schueler
cur.execute("""
CREATE TABLE Schueler (
    SNr INTEGER PRIMARY KEY,
    Name TEXT NOT NULL,
    Vorname TEXT NOT NULL
)
""")

# Tabelle LKWahl
cur.execute("""
CREATE TABLE LKWahl (
    SNr INTEGER,
    KID TEXT,
    PRIMARY KEY (SNr, KID),
    FOREIGN KEY (SNr) REFERENCES Schueler(SNr),
    FOREIGN KEY (KID) REFERENCES Kurse(KID)
)
""")

# Daten einfügen

# Lehrer
cur.execute("INSERT INTO Lehrer VALUES ('SC', 'Schulte')")
cur.execute("INSERT INTO Lehrer VALUES ('DA', 'Dallmeyer')")
cur.execute("INSERT INTO Lehrer VALUES ('AD', 'Adam')")
cur.execute("INSERT INTO Lehrer VALUES ('GE', 'Gebauer')")

# Kurse
cur.execute("INSERT INTO Kurse VALUES ('M1', 'SC')")
cur.execute("INSERT INTO Kurse VALUES ('Ph1', 'SC')")
cur.execute("INSERT INTO Kurse VALUES ('Bio1', 'DA')")
cur.execute("INSERT INTO Kurse VALUES ('M2', 'AD')")
cur.execute("INSERT INTO Kurse VALUES ('Ge2', 'GE')")

# Schüler
cur.execute("INSERT INTO Schueler VALUES (120, 'Riedberg', 'Malte')")
cur.execute("INSERT INTO Schueler VALUES (121, 'Dreis', 'Lena')")
cur.execute("INSERT INTO Schueler VALUES (122, 'Rödding', 'Maike')")

# LK-Wahlen
cur.execute("INSERT INTO LKWahl VALUES (120, 'M1')")
cur.execute("INSERT INTO LKWahl VALUES (120, 'Ph1')")
cur.execute("INSERT INTO LKWahl VALUES (121, 'Bio1')")
cur.execute("INSERT INTO LKWahl VALUES (121, 'Ph1')")
cur.execute("INSERT INTO LKWahl VALUES (122, 'M2')")
cur.execute("INSERT INTO LKWahl VALUES (122, 'Ge2')")

con.commit()

print("Datenbank erfolgreich erstellt.")

con.close()
```

**A3** Gesucht sind alle Kurse von Schulte oder Adam



```python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT KID, Lehrer
    FROM Kurse, Lehrer
    WHERE Kurse.Kuerzel = Lehrer.Kuerzel
      AND (Lehrer = 'Schulte'
           OR Lehrer = 'Adam')          
""")
for datensatz in cur.fetchall():
    print(datensatz)

con.close()
```

Schema: <br> 
LKWahl(<u>SNr</u>↑,<u>KID</u>↑) <br> 
Schueler(<u>SNr</u>,Name,Vorname) <br> 
Kurse(<u>KID</u>,Kuerzel↑) <br> 
Lehrer(<u>Kuerzel</u>,Lehrer)

**A4** Gib für jede LK-Wahl den Namen des Schülers bzw. der Schülerin, die KursID und den Namen der Lehrkraft aus, aufsteigend sortiert nach Nachnamen.

```python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT Name, Vorname, LKWahl.KID, Lehrer
    FROM Schueler, LKWahl, Kurse, Lehrer
    WHERE Schueler.SNr = LKWahl.SNr
      AND LKWahl.KID = Kurse.KID
      AND Kurse.Kuerzel = Lehrer.Kuerzel
    ORDER BY Name 
""")

for datensatz in cur.fetchall():
    print(datensatz)

con.close()
```

Für formatierte Ausgaben können wir unpacking und f-String benutzen

```python
import sqlite3

con = sqlite3.connect("lkwahl.db")
cur = con.cursor()

cur.execute("""
    SELECT Name, Vorname, LKWahl.KID, Lehrer
    FROM Schueler, LKWahl, Kurse, Lehrer
    WHERE Schueler.SNr = LKWahl.SNr
      AND LKWahl.KID = Kurse.KID
      AND Kurse.Kuerzel = Lehrer.Kuerzel
    ORDER BY Name
""")

for name, vorname, kid, lehrer in cur.fetchall():          # unpacking
    print(f'{vorname:10} {name:10} {kid:4} {lehrer:10}')   

con.close()
```
