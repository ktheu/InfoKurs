#### Scenario 1

Wir machen von Zeit zu Zeit Sicherungen um bei Bedarf auf ältere Stände zurückgreifen zu können.

<img src='git01.png' width='700'>

- wd: Working directory  = der aktuelle Code <br>
- sa: Staging area  = der Warenkorb für den nächsten Commit <br>
- co: Commits  = der Bereich der gesicherten Stände <br>

Ablauf: wir editieren in unserem working directory, bringen alles, was in die nächste Sicherung soll in die staging area und sichern dann mit einem commit.



Bei add und commit bekommen wir Hinweise, wie wir etwas rückgängig machen können.

<img src='git02.png' width='900'>

```python
git restore f1.txt                         # nimmt Änderungen im working dir zurück, falls nicht gestaged
git restore --staged f1.txt                # nimmt Datei aus staging area zurück
git restore --staged --workingtree f1.txt  # nimmt Datei aus staging area zurück und nimmt auch die Änderungen im working directory zurück
```

#### checkout: Head im detached mode

Wenn wir den Head auf einen früheren commit zeigen lassen, können wir uns diesen Zustand ansehen ohne den aktuellen Zustand zu verlieren. Der HEAD befindet sich dann im 'detached'-Mode, weil er nicht mehr auf einen branch zeigt.

<img src='git04.png' width='700'>

```python
git checkout HEAD~1             # HEAD auf vorletzten commit
git checkout main               # HEAD auf branch main
git checkout -                  # HEAD auf den branch, auf den er zuletzt war
```

#### git stash

Den Ausflug in eine frühere Welt mit dem detached HEAD können wir nur machen, wenn der working-tree clean ist, d.h. wir haben keine Änderungen, die noch nicht committed sind. Wenn doch Änderungen da sind, müssen wir die in einen getrennten Bereich *verstauen* (stash) und können sie nach dem Ausflug wieder zurückholen.

<img src='git05.png' width='701'>

#### git diff

Wir können unser working-directory mit früheren Zuständen vergleichen

```python
git diff HEAD~1 f1.txt
git diff 096987e  
```

#### Scenario 2
Wir haben einen Stand, der in Gebrauch ist und den wir nicht durch Experimente unbrauchbar machen wollen. Trotzdem wollen wir darauf aufbauend ein neues Feature entwickeln. Dazu nutzen wir *branches*.

<img src='git06.png' width='600'>

```python
git switch -c feature         # neuen branch erstellen
git switch main               # branch wechseln
git branch                    # list der branches 
```

Wir können einen neuen branch auch an einen früheren commit ansetzen. Dazu versetzen wir den HEAD kurzzeitig in den detached mode.

<img src='git10.png' width='450'>

Wenn wird die Änderungen im branch feature in den main-branch übernehmen wollen, machen wir einen merge. In dieser Situation wird nur der Head des masters versetzt, es wird kein neuer commit gemacht. Dies ist ein *fast forward merge*

<img src='git06.png' width='600'>

#### Rebase

Auch mit Rebase können die Änderungen zweier branches zusammenengeführt werden. 

<img src='git11.png' width='900'>

Auch beim Rebase können Konflikte entstehen, die gelöst werden müssen. Die Commits für c4 haben zwar den gleichen Inhalt, aber es sind verschiedene Commits mit unterschiedlichen Hashwerten. Rebase sollte nicht benutzt werden in öffentlichen Repos, an denen auch andere mitarbeiten.

#### git cherry-Pick

Wir holen uns aus einem anderen branch einen speziellen Commit heraus und fügen den in den aktuellen branch ein.

```python
git cherry-pick 421512c
```

#### git reset

Mit *git reset* können wir zu einem commit zurückgehen und die nachfolgenden commits löschen, ohne die Änderungen im working directory zu verlieren. Wir können dann die Änderungen z.B. in einen anderen branch einfügen.

<img src='git12.png' width='500'>

Mit *git reset --hard* setzen wir auch die Änderungen im working directory zurück.

<img src='git13.png' width='500'>

#### git revert

Auch mit git *revert* kann ich auf den Zustand eines älteren commits zurückgehen, aber nicht durch Löschen von commits, sondern durch Rücknahme der Änderungen in einem neuen commit.

<img src='git14.png' width='500'>

#### .gitignore

Ein .gitignore-File in dem lokalen Repo sorgt dafür, dass bestimmte Dateien nicht mit Git verwaltet werden. 
.gitignore beeinflusst nur nicht getrackte Dateien.

[Muster für .gitignore-Files](https://github.com/github/gitignore)

```python
# Kommentare beginnen mit #
*.log              # Log-Dateien ignorieren
geheim.txt         # diese Datei ignorieren
notizen/           # Notizen-Ordner ignorieren
```

Ist eine Datei schon getracked und wollen wir sie trotzdem ignorieren, dann müssen wir sie erst aus der git-Verwaltung herausnehmen.

```python
git rm --cached geheim.txt
```
