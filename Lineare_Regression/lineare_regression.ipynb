### Lineare Regression

```python
data = [(1,2),(2,4),(3,5),(4,4),(5,5)]
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data points
x = np.array([d[0] for d in data])
y = np.array([d[1] for d in data])

# Regression coefficients 
a = 0.6
b = 2.2

# Create line for regression
x_line = np.linspace(0, 6, 100)
y_line = a * x_line + b

# Plot
plt.scatter(x, y, color="blue", label="Data points")
plt.plot(x_line, y_line, color="red", label=f"y = {a}x + {b}")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```

#### Rechenweg mit sympy:

```python
import sympy as sp
```

```python
a, b = sp.symbols("a b")
```

Die Summe der quadratischen Abweichungen bilden

```python
summe = 0
for x, y in data:
    summe = summe + (a * x + b - y)**2
summe
```

Die partiellen Ableitungen nach a und b bilden.

```python
g1 = sp.diff(summe,a)
g1
```

```python
g2 = sp.diff(summe,b)
g2
```

Das Gleichungssystem lösen

```python
sp.solve((g1, g2), (a, b))
```

#### scipy

Mit scipy können wir uns den Achsenabschnitt und die Steigung der Regressionsgeraden berechnen lassen.

```python
from scipy import stats

# Daten
x = np.array([d[0] for d in data])
y = np.array([d[1] for d in data])

# Lineare Regression
result = stats.linregress(x, y)
print("Steigung a:", result.slope)
print("Achsenabschnitt b:", result.intercept)
```
