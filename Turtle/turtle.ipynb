### Rekursive Bilder mit der Turtle

[doc](https://docs.python.org/3/library/turtle.html#turtle-motion) - [video1](https://youtu.be/57-1jupLbAA) - [video2](https://youtu.be/WaZedLeexS8)

Turtle: In Thonny öffnet sich ein Fenster. Der 0-Punkt ist in der Mitte. Breite und Höhe sind 768 x 648.

```python
from turtle import *
forward(100)          # 100 Pixel vorwärts = fd(100)
backward(100)
right(90)             # drehe nach rechts um 90 Grad = rt(90)
left(90)              # lt(90)

up()                  # Stift anheben
down()                # Stift absetzen
hideturtle()          # Turtle verbergen = ht()

home()                # zum Ursprung (0/0) gehen
goto(200,100)         # Turtle an die Position bewegen
position()            # Position der Turtle: (x,y)

heading()             # Richtung der Turtle: 0 = nach rechts, 90 = nach oben ..., d.h. Winkel gegen den Uhrzeigersinn
setheading(270)       # Richtung der Turtle nach unten setzen.

pensize(5)            # setze Stiftstärke auf 5
color('blue')         # setze Stiftfarbe und Füllfarbe

circle(50)            # Kreis mit Radius 50    
dot(30,'red')         # roter Punkt mit Durchmesser 30
write('Aha')          # String an die aktuelle Turtle-Position schreiben    

speed(0)              # schnell
tracer(0,0)           # ohne Animation
```

#### Rekursive Linien

Wir geben die Anzahl der Rekursionen mit der Variablen bremse vor.

<img src='linien.png' width='600'>

```python
from turtle import *

def zeichne(laenge, bremse):
    if bremse == 0:
        return
    down()
    forward(laenge)
    up()
    forward(laenge//2)
    zeichne(laenge*0.85, bremse-1)

hideturtle()
speed(0)
up()
goto(-300,0)

zeichne(60, 20)
```

#### Rekursive Verzweigungen

<img src='verzweigungA.png' width='401'>

```python
# Mit Reihenfolge, in der die Rekursionsaufrufe starten
from turtle import *
    
def zeichne(laenge,bremse):
    if bremse == 0: return
    
    # Ausgangsposition merken
    pos = position()
    winkel = heading()
   
    # 1. Verzweigung
    right(40)
    forward(laenge)
    zeichne(laenge*0.7,bremse-1)
    
    # zurück auf Ausgangsposition
    up()
    setpos(pos)
    setheading(winkel)
    down()
    
    # 2. Verzweigung
    left(40)
    forward(laenge)
    zeichne(laenge*0.7,bremse-1)
    
hideturtle()
pensize(2) 
speed(0)
up()
goto(0,-200)
down()
left(90)
zeichne(150,2)
```

Wir können die Reihenfolge ausgeben, in der die Rekursionsaufrufe starten.

<img src='verzweigung_nr.png' width='600'>

```python
from turtle import *
    
def zeichne(laenge,bremse):
    global zaehl
    zaehl+=1
    write(zaehl)
    if bremse == 0:
        dot(8,'green')
        return
    dot(8,'red')  
    
    # Ausgangsposition merken
    pos = position()
    winkel = heading()
    
    # 1. Verzweigung
    right(40)
    forward(laenge)
    zeichne(laenge*0.7,bremse-1)
    
    # zurück auf Ausgangsposition
    up()
    setpos(pos)
    setheading(winkel)
    down()
  
    
    # 2. Verzweigung
    left(40)
    forward(laenge)
    zeichne(laenge*0.7,bremse-1)
 
    
hideturtle()
zaehl = 0
pensize(1)
speed(0)
up()
goto(0,-200)
down()
left(90)
zeichne(150,3)
```

Wir können auch die Reihenfolge ausgeben, in der die Rekursionsaufrufe enden.

<img src='verzweigung_ende.png' width='600'>

```python
# Mit Reihenfolge, in der die Rekursionsaufrufe enden
from turtle import *
    
def zeichne(laenge,bremse):
    global zaehl

    if bremse == 0:
        dot(8,'green')
        zaehl+=1
        write(zaehl)
        return
    dot(8,'red')  
    
    # Ausgangsposition merken
    pos = position()
    winkel = heading()
    
    # 1. Verzweigung
    right(40)
    forward(laenge)
    zeichne(laenge*0.7,bremse-1)

    # zurück auf Ausgangsposition
    up()
    setpos(pos)
    setheading(winkel)
    down()
  
    
    # 2. Verzweigung
    left(40)
    forward(laenge)
    zeichne(laenge*0.7,bremse-1)
    
    
    up()
    setpos(pos)
    setheading(winkel)
    down()
    zaehl+=1
    write(zaehl)
 
    
hideturtle()
zaehl = 0
pensize(1)
speed(0)
up()
goto(0,-200)
down()
left(90)
zeichne(150,3)
```

#### Dreiecke

<img src='dreiecke.png' width='301'>

```python
from turtle import *

def dreieck(laenge,bremse):
    if bremse == 0: return
    
    left(90)
    fd(laenge//2)

    for i in range(3):
        pos = position()
        head = heading()
        
        left(30)
        dreieck(0.45*laenge,bremse-1)
        
        setpos(pos)
        setheading(head)
        right(120)

        if i == 2:
            fd(laenge//2)
        else:
            fd(laenge)

farben = ['red','green','blue','orange']
hideturtle()
speed(0)
up()
goto(0,-100)
down()
left(90)
dreieck(200,2)
```

Wir können die Rekursionsstufen auch farblich kennzeichnen

<img src='dreiecke_farben.png' width='500'>

```python
from turtle import *

def dreieck(laenge,bremse):
    if bremse == 0: return
    
    left(90)
    color(farben[bremse%len(farben)])
    fd(laenge//2)

    for i in range(3):
        pos = position()
        head = heading()
        
        left(30)
        dreieck(0.45*laenge,bremse-1)
        
        setpos(pos)
        setheading(head)
        right(120)
        
        color(farben[bremse%len(farben)])
        if i == 2:
            fd(laenge//2)
        else:
            fd(laenge)
  
farben = ['red','green','blue','orange','yellow']  
hideturtle()
speed(0)
up()
goto(0,-100)
down()
left(90)
pensize(2)
dreieck(200,4)
```

#### Sierpinski Dreieck

Nur bei der Rekursionsbremse wird gezeichnet. Die Rekursion dient dazu, die Positionen zu finden.

<img src='sierpinski.png' width='500'>

Das Muster:

<img src='sierpinski0.png' width='500'>

```python
from turtle import *
    
def zeichne(laenge,bremse):
    
    if bremse == 0:
        down()
        for i in range(3):    # Dreieck zeichnen
            forward(laenge)
            left(120)
        up()
        return
    
    pos = position()
    winkel = heading()
    
    zeichne(laenge//2,bremse-1)    # 1. Dreieck links unten
    
    forward(laenge//2)
    zeichne(laenge//2,bremse-1)    # 2. Dreieck rechts unten
    
    left(120)
    forward(laenge//2)
    right(120)
    
    zeichne(laenge//2,bremse-1)
    
    setpos(pos)
    setheading(winkel)

tracer(0,0)
hideturtle()
up()
goto(-320,-280)
zeichne(650,4)

```

#### Kochsche Schneeflocke

Nur bei der Rekursionsbremse wird gezeichnet. Die Rekursion dient dazu, die Position und Winkel zu finden.

<img src='koch.png' width='500'>

Das Muster:

<img src='koch_muster.png' width='700'>

```python
# Eine Seite der Kochschen Schneeflocke
from turtle import *
    
def zeichne(laenge,bremse):
    if bremse == 0: 
        down()
        forward(laenge)
        up()
        return
    
    zeichne(laenge//3,bremse-1)
    left(60)
    zeichne(laenge//3,bremse-1)
    right(120)
    zeichne(laenge//3,bremse-1)
    left(60)
    zeichne(laenge//3,bremse-1)

#hideturtle()
speed(4)
up()
goto(-200,0)
zeichne(400,1)
```

Alle drei Seiten zusammen:

```python
from turtle import *
    
def zeichne(laenge,bremse):
    if bremse == 0: 
        down()
        forward(laenge)
        up()
        return
    
    drittel = laenge/3   
    
    zeichne(drittel,bremse-1)
    left(60)
    zeichne(drittel,bremse-1)
    right(120)
    zeichne(drittel,bremse-1)
    left(60)
    zeichne(drittel,bremse-1)

hideturtle()
tracer(0,0)
up()
goto(-300,150)

for i in range(3):    # 3 Seiten der Schneeflocke
    zeichne(500,3)
    right(120)
 
```
