### Rekursive Bilder mit der Turtle

Turtle: In Thonny öffnet sich ein Fenster. Der 0-Punkt ist in der Mitte. Breite und Höhe sind 768 x 648.

```python
from turtle import *
forward(100)          # 100 Pixel vorwärts = fd(100)
backward(100)
right(90)             # drehe nach rechts um 90 Grad = rt(90)
left(90)              # lt(90)

up()                  # Stift anheben
down()                # Stift absetzen
hideturtle()          # Turtle verbergen = ht()

home()                # zum Ursprung (0/0) gehen
goto(200,100)         # Turtle an die Position bewegen
position()            # Position der Turtle: (x,y)

heading()             # Richtung der Turtle: 0 = nach rechts, 90 = nach oben ..., d.h. Winkel gegen den Uhrzeigersinn
setheading(270)       # Richtung der Turtle nach unten setzen.

pensize(5)            # setze Stiftstärke auf 5
color('blue')         # setze Stiftfarbe und Füllfarbe

circle(50)            # Kreis mit Radius 50    
dot(30,'red')         # roter Punkt mit Durchmesser 30
write('Aha')          # String an die aktuelle Turtle-Position schreiben    

speed(0)              # schnell
tracer(0,0)           # ohne Animation
```

#### Rekursive Linien

Wir geben die Anzahl der Rekursionen mit der Variablen bremse vor.

<img src='linien.png' width='600'>

```python
from turtle import *

def zeichne(laenge, bremse):
    if bremse == 0:
        return
    down()
    forward(laenge)
    up()
    forward(laenge//2)
    zeichne(laenge*0.85, bremse-1)

hideturtle()
speed(0)
up()
goto(-300,0)

zeichne(60, 20)
```

#### Rekursive Quadrate

<img src='quadrate.png' width='400'>

```python
from turtle import *
    
def zeichne(laenge, bremse):
    if bremse == 0:
        return
    fd(laenge)
    left(90)
    zeichne(laenge*0.98, bremse-1)

up()
goto(-200,-200)
down()
pensize(1);
hideturtle()
speed(0)
zeichne(400,100)
```

#### Rekursive Verzweigungen

<img src='verzweigung.png' width='400'>

```python
from turtle import *
    
def zeichne(laenge,bremse):
    if bremse == 0:
        return
    
    # Ausgangsposition merken
    pos = position()
    winkel = heading()
    
    # 1. Verzweigung
    right(45)
    forward(laenge)
    zeichne(laenge*0.5,bremse-1)
    
    # zurück auf Ausgangsposition
    up()
    setpos(pos)
    setheading(winkel)
    down()
    
    # 2. Verzweigung
    left(45)
    forward(laenge)
    zeichne(laenge*0.5,bremse-1)
    
hideturtle()
 
pensize(3)
speed(10)
left(90)
zeichne(200,4)
 
```

#### Sierpinski Dreieck

Nur bei der Rekursionsbremse wird gezeichnet. Die Rekursion dient dazu, die Positionen zu finden.

<img src='sierpinski.png' width='500'>

```python
from turtle import *
    
def zeichne(laenge,bremse):
    
    if bremse == 0:
        down()
        for i in range(3):    # Dreieck zeichnen
            forward(laenge)
            left(120)
        up()
        return
    
    halb = laenge/2
    
    pos = position()
    winkel = heading()
    
    zeichne(halb,bremse-1)    # 1. Dreieck links unten
    
    forward(halb)
    zeichne(halb,bremse-1)    # 2. Dreieck rechts unten
    
    left(120)
    forward(halb)
    right(120)
    
    zeichne(halb,bremse-1)
    
    setpos(pos)
    setheading(winkel)
 
 
tracer(0,0)
up()
goto(-350,-280)
hideturtle()
zeichne(650,4)
```

#### Kochsche Schneeflocke

Nur bei der Rekursionsbremse wird gezeichnet. Die Rekursion dient dazu, die Position und Winkel zu finden.

<img src='koch.png' width='500'>

```python
from turtle import *
    
def zeichne(laenge,bremse):
    if bremse == 0: 
        down()
        forward(laenge)
        up()
        return
    
    drittel = laenge/3   
    
    zeichne(drittel,bremse-1)
    left(60)
    zeichne(drittel,bremse-1)
    right(120)
    zeichne(drittel,bremse-1)
    left(60)
    zeichne(drittel,bremse-1)

hideturtle()
tracer(0,0)
up()
goto(-300,150)

for i in range(3):    # 3 Seiten der Schneeflocke
    zeichne(500,3)
    right(120)
 
```
