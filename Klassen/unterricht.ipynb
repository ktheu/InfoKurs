```python
''' Kontostand
Erstelle eine Klasse Konto  
- Attribute: inhaber, stand
- Methoden: einzahlen, abheben

Beispiel:

k = Konto("Mia", 100)
k.einzahlen(50)
print(k)
k.abheben(70)
print(k)
k.abheben(100)
print(k)

Erwartete Ausgabe:
Kontostand: 150
Kontostand: 80
Nicht genug Geld!
Kontostand: 80
'''
class Konto:
    def __init__(self, inhaber, stand):
        self.inhaber = inhaber
        self.stand = stand

    def einzahlen(self,geld):
        self.stand = self.stand + geld


    def abheben(self,geld):
        if self.stand - geld < 0:
            print('Nicht genug Geld!')
        else:
            self.stand = self.stand - geld

    def __str__(self):
        return f'Kontostand: {self.stand}'
        

    
    
```

```python
k = Konto("Mia", 100)
k.einzahlen(50)
print(k)
k.abheben(70)
print(k)
k.abheben(100)
print(k)
```

```python
'''
Erstelle die Klasse Buch  
- Attribute: titel, autor, jahr  

Erstelle die Unterklasse Schulbuch  
- Attribut: fach  

Beispiel:

b = Buch("Der Hobbit", "Tolkien", 1937)
s = Schulbuch("Algebra", "Müller", 2020, "Mathematik")
print(b)
print(s)

Erwartete Ausgabe:

Der Hobbit von Tolkien (1937)
Fach: Mathematik, Algebra von Müller (2020)
'''
class Buch:
    def __init__(self, titel, autor, jahr):
        self.titel = titel
        self.autor = autor
        self.jahr = jahr


    def __str__(self):
        return f'{self.titel} von {self.autor} ({self.jahr})'
        
class Schulbuch(Buch):
    def __init__(self, titel, autor, jahr, fach):
        super().__init__(titel, autor, jahr)
        self.fach = fach

    def __str__(self):
        return f'Fach: {self.fach}, ' + super().__str__()
```

```python
b = Buch("Der Hobbit", "Tolkien", 1937)
s = Schulbuch("Algebra", "Müller", 2020, "Mathematik")
print(b)
print(s)
```
