##### Steve

Steve ist sehr schüchtern und zurückgezogen, immer hilfsbereit, aber mit sehr geringem Interesse
an Menschen oder an der realen Welt.
Eine sanfte und ordentliche Seele, er hat ein Bedürfnis nach Ordnung
und Struktur sowie eine Leidenschaft für Details. Was hältst du für wahrscheinlicher:
Steve ist Bibliothekar oder Steve ist Landwirt?

Wir glauben, dass es ca. 30 mal mehr Landwirte als Bibliothekare gibt (**Prior**). Wir schätzen, dass die Beschreibung von Steve auf ca. 30% der Bibliothekare und 4% der Landwirte passt. (**Likelihood**)

Mit der **Formel von Bayes** können wir die Wahrscheinlichkeit ausrechnen, dass es sich bei Steve um einen Bibliothekar handelt.
Das Verfahren wird manchmal mit den Worten beschrieben: **update belief based on new evidence**.
